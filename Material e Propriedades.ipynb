```python
# Yield

def _yield(n):
    soma = 0
    for i in range(1, n+1):
        soma += i
    yield soma
    
soma_total = 0
for parcial in _yield(9):
    soma_total = parcial
print(soma_total)
```

```python
# FUNCTOOLS
import functools

@functools.cache # Armazena as resposta na memória cache. assim, ele não precisará recalcular
def fatorial(n):
    print(f'Valor de n: {n}')
    if n == 1:
        return n
    return n * fatorial(n-1)
fatorial(4)
```

```python
# Herança
class Registro:
    def __init__(self, nome, idade, letra, codigo):
        self.nome = nome
        self.idade = idade
        self.letra = letra
        self.__codigo = codigo
    
    def aniversario(self, acrescentar):
        self.idade += acrescentar
        return self.idade
        
    def alterar_letra(self, alterar):
        self.letra = alterar
        return self.letra
    
    def alterar_senha(self, senha):
        self.__codigo = senha
    
    def resultado(self):
        return self.__codigo
    
```

```python
class Loja:
    def __init__(self, nome, preco, quantidade):
        self.nome = nome
        self.preco = preco
        self.quantidade = quantidade
        
    def __str__(self):
        return f'{self.nome} | R${self.preco} | {self.quantidade}'
    
         
    
class Vendedor(Loja):
    def __init__(self, nome, preco, quantidade, vendedor, garantia):
        super().__init__(nome, preco, quantidade)
        self.vendedor = vendedor
        self.garantia = garantia
        
    def __str__(self):
        return f'{self.nome} | {self.preco} | {self.quantidade} | {self.vendedor} | {self.garantia} '
        
        
class Gerenciamento(Vendedor):
    def __init__(self):
        self.produtos = []
        
    def adicionar(self, produto):
        self.produtos.append(produto)
    
    def exibir(self):
            for produto in self.produtos:
                print(produto)
        
gerenciador= Gerenciamento()

obj1 = Vendedor('Carro', '45000', '4', 'Victor', 'Sim')
gerenciador.adicionar(obj1)
type(obj1)
obj1 = Gerenciamento()
type(obj1)
```

```python
class Comp:
    def __init__(self, deslocamento, tempo):
        self.deslocamento = deslocamento
        self.tempo = tempo
        
    def velocidade(self):
        self.velocidade = self.deslocamento / self.tempo
        print(f"A velocidade é {self.velocidade} m/s")

class Velo(Comp):
    def __init__(self, deslocamento, tempo, deslextra):
        super().__init__(deslocamento, tempo)
        self.deslextra = deslextra
        
    def velocidade(self):
        super().velocidade()
        self.veloextra = self.deslextra / self.tempo
        self.diff = ((self.veloextra - self.velocidade)**2)**0.5
        print(f'A velocidade extra é {self.veloextra} m/s')
        print(f'A diferença de velocidade é {self.diff} m/s')
        
        
v = Velo(15, 5, 10)
v.velocidade()
```

```python
# Decorador

import time

def decorador(func):
    def wrapper():
        print('Inicio da contagem')
        inicio = time.perf_counter()
        resultado = func()
        fim = time.perf_counter()
        print('Final da contagem')
        tempo_decorrido = fim - inicio
        print(f'{tempo_decorrido: .15f} segundos')
    return wrapper
              
@ decorador
def velo(deslocamento = 10, tempo = 2):
              velocidade = deslocamento / tempo
              print(f"A velocidade é {velocidade} m/s")
@ decorador
def acele(deslocamento = 10, tempo = 2, tempo1 = 3):
              aceleracao = deslocamento / (tempo*tempo1)
              print(f"A aceleração é {aceleracao: .3f} m/s²")
velo()
acele()
```

```python
# Melhorando o desempenho da classe reduzindo o o impacto da memória com __slot__ ao invés de um dict(default)

class Emply:
    __slot__ = ['nome', 'idade', 'supri']
    def __init__(self,nome,idade):
        self.nome = nome
        self.idade = idade
    def __repr__(self):
        return f'Nome: {self.nome} Idade: {self.idade}'
    def __str__(self):
        return f'Nome: {self.nome} Idade: {self.idade}'
    
e = Emply('Victor', 28)
e.supri = 1000 # 
print(f'({e.nome}, {e.idade}, {e.supri})')
```

```python
# Override e polimorfismo
# Override: Criar métodos com o mesmo nome para classes diferentes
# Polimorfismo: Com os métodos do override, criar funções diferentes entre as classes

class Carro:
    def __init__(self, nome, cor, chassi):
        self.nome = nome
        self.cor = cor
        self.chassi = chassi
        
    def __repr__(self):
        return f'Nome: {self.nome}  Cor: {self.cor}  Chassi: {self.chassi}'
    
    def __str__(self):
        return f'Nome: {self.nome}  Cor: {self.cor}  Chassi: {self.chassi}'
    
    def a(self):
        return f'Nome: {self.nome}  Cor: {self.cor}  Chassi: {self.chassi}'


class Objeto(Carro):
    def __init__(self, nome, cor, chassi, marca):
        super().__init__(nome, cor, chassi) # Super só no init da superclasse
        self.marca = marca
        
    def a(self): 
        super().a() # Override e polimorfismo
        return f'Nome: {self.nome}  Cor: {self.cor}  Chassi: {self.chassi}   Marca: {self.marca}'

# Executar o código diretamente
if __name__ == "__main__":        
    c = Carro('Uno', 'Cinza', 182930)
    o = Objeto('Uno', 'Cinza', 182930, 'Fiat')
    print(c.a())
    print(o.a())
else:
    print('Esse script foi importado como módulo')
```

```python
class Person:
    def __init__(self, first_name):
        self.first_name = first_name
# Getter function
@property
def first_name(self):
    return self._first_name
# Setter function
@first_name.setter
def first_name(self, value):
    if not isinstance(value, str): # Necessário ser uma string
        raise TypeError('Expected a string')
    self._first_name = value
# Deleter function (optional)
@first_name.deleter
def first_name(self):
    raise AttributeError("Can't delete attribute")
```

```python
# POLIMORFISMO 

# TODAS AS CLASSES RECEBERÃO TODOS OS MÉTODOS IDENTICOS, SE NÃO, DARÁ UM ERRO

from abc import ABC, abstractmethod # 

class Animal(ABC):
    @abstractmethod
    def fazer_som(self):
        pass

class Cachorro(Animal):
    def fazer_som(self):
        return "Au Au"

class Gato(Animal):
    def fazer_som(self):
        return "Miau"

class Porco(Animal):
    def fazer_som(self):
        return "Hoc Hoc"

# Tentativa de instanciar a classe abstrata resultará em erro
# animal = Animal()  # Isso gera um erro

# Instanciando subclasses
cachorro = Cachorro()
gato = Gato()
porco = Porco()
print(cachorro.fazer_som())  # Saída: Au Au
print(gato.fazer_som())      # Saída: Miau
print(porco)  
```
