# Camada de Valência
- Camada mais externa que orbita o núcleo da partícula.
# Camada de Condução
- Camada que conduz os **elétrons livres** que viajam entre um átomo à outro.
  - elétrons livres: São elétrons que contém pouca atração com o núcleo por estarem distântes do núcleo.
# Fenda Proibida
- Região onde o elétron precisa passar para ir ao átomo vizinho.
# Isoladores
- Possuem uma grande Fenda Proibida
- A lacuna de energia proibida é muito grande
- Os elétrons da banda de valência estão fortemente ligados aos átomos.
- O valor da lacuna de energia proibida para um isolante será de 10 eV.
- Para alguns isolantes, à medida que a temperatura aumenta, eles podem apresentar alguma condutividade.
- A resistividade de um isolante será da ordem de $10^7$ ohm-metros.
# Semicondutores
- A lacuna de energia proibida é uma pequena.
- A banda proibida para o germânio (Ge) é de 0,7 eV, enquanto que para o silício (Si) é de 1,1 eV.
- Um semicondutor, na verdade, não é nem um isolante, nem um bom condutor.
- Com o aumento da temperatura, a condutividade de um semicondutor aumenta.
- A condutividade de um semicondutor será da ordem de $10^2$ mho-metros.
# Condutores
- Não existe uma lacuna proibida.
- A camada de valência e a camada de condução se sobrepõem.
- Os elétrons livres disponíveis para condução são abundantes.
- Um ligeiro aumento na voltagem aumenta a condução.
- Não existe o coneito de formação de lacunas, pois um fluxo contínuo de elétrons contribui para a corrente


Semicondutores é uma substância cuja resistividade se situa entre a de condutores e a de isolantes.
- Os semicondutores possuem resistividade menor que a dos isolantes e maior que a dos condutores.
- Os semicondutores possuem coeficiente de temperatura negativo. A resistência em semicondutores<br>
aumenta com a diminuição da temperatura e vice-versa
- As propriedades condutoras de um semicondutor se alteram quando uma impureza metálica adequada <br>
é adicionada a ele, o que é uma propriedade muito importante.



# Condução em semicondutores 
Quando um átomo com elétrons de valência é aproximado de outro átomo, <br>
os elétrons de valência de ambos se combinam para formar pares de elétrons. <br>
Essa ligação não é muito forte e, portanto, é uma ligação covalente. <br>

# Criação de um buraco 
Devido à energia térmica (calor "Q") fornecida ao cristal, alguns elétrons tendem a se deslocar de seus lugares e <br>
romper as ligações covalentes. Essas ligações covalentes rompidas resultam em elétrons livres que vagam <br>
aleatoriamente. Mas os elétrons deslocados criam um espaço vazio ou camada de valência atrás, que é chamado <br>
de buraco. <br>
$Ge + Q \implies Ge^+ + E$

Essa lacuna, que representa um elétron ausente, pode ser considerada como uma unidade de carga positiva, enquanto <br>
o elétron é considerado como uma unidade de carga negativa. Os elétrons liberados movem-se aleatoriamente, mas quando <br>
um campo elétrico externo é aplicado, esses elétrons movem-se na direção oposta à do campo aplicado. Já as lacunas criadas<br>
pela ausência de elétrons movem-se na direção do campo aplicado. <br>

# Corrente de buraco 
Já se sabe que, quando uma ligação covalente é quebrada, cria-se uma lacuna. Na verdade, existe uma <br> 
forte tendência dos cristais semicondutores em formar ligações covalentes. Portanto, não é comum a <br> 
existência de lacunas em um cristal



Quando um elétron é deslocado de um ponto A, forma-se uma lacuna. Devido à tendência de formação de ligações covalentes, <br>
um elétron de B é deslocado para A. Novamente, para equilibrar a ligação covalente em B, um elétron é deslocado de C para B. <br>
Isso continua a construir um caminho. Esse movimento da lacuna, na ausência de um campo elétrico aplicado, é aleatório. Mas<br>
quando um campo elétrico é aplicado, a lacuna se desloca ao longo do campo, o que constitui a corrente de lacunas . Essa<br>
corrente é chamada de corrente de lacunas, e não de corrente de elétrons, porque o movimento das lacunas contribui para o <br>
fluxo de corrente. 

Elétrons e lacunas, em movimento aleatório, podem se encontrar e formar pares. Essa recombinação resulta na liberação de calor, <br>
que rompe outra ligação covalente. Quando a temperatura aumenta, a taxa de geração de elétrons e lacunas aumenta, assim como a <br>
taxa de recombinação, o que resulta no aumento da densidade de elétrons e lacunas. Consequentemente, a condutividade do <br>
semicondutor aumenta e a resistividade diminui, o que significa um coeficiente de temperatura negativo. 

# Semicondutores intrínsecos 
Um semicondutor em sua forma extremamente pura é chamado de semicondutor intrínseco. <br>
As propriedades desse semicondutor puro são as seguintes: 
- Os elétrons e lacunas são criados exclusivamente por excitação térmica.
- O número de elétrons livres é igual ao número de lacunas.
- A capacidade de condução é pequena à temperatura ambiente.

Para aumentar a capacidade de condução de um semicondutor intrínseco, é recomendável adicionar impurezas. <br>
Esse processo de adição de impurezas é chamado de dopagem . O semicondutor intrínseco dopado resultante é <br>
denominado semicondutor extrínseco. 

# Doping 
O processo de adição de impurezas aos materiais semicondutores é denominado dopagem. As impurezas adicionadas <br>
são geralmente impurezas pentavalentes e trivalentes. 

1. Impurezas pentavalentes
- As impurezas pentavalentes são aquelas que possuem cinco elétrons de valência na camada mais externa. Exemplos: <br>
bismuto, antimônio, arsênio, fósforo. 
- O átomo pentavalente é chamado de átomo doador porque doa um elétron para a banda de condução de um átomo semicondutor <br>
puro. 

2. Impurezas trivalentes
- As impurezas trivalentes são aquelas que possuem três elétrons de valência na camada mais externa. Exemplos: Gálio, Índio, <br>
Alumínio, Boro.
- O átomo trivalente é chamado de átomo aceptor porque aceita um elétron do átomo semicondutor.



# Semicondutor extrínseco 
Um semicondutor impuro, formado pela dopagem de um semicondutor puro, é chamado de semicondutor <br>
extrínseco . Existem dois tipos de semicondutores extrínsecos, dependendo do tipo de impureza <br>
adicionada: semicondutores extrínsecos do tipo N e semicondutores extrínsecos do tipo P. 

# Semicondutor extrínseco do tipo N 
Uma pequena quantidade de impureza pentavalente é adicionada a um semicondutor puro, resultando em um semicondutor <br>
extrínseco do tipo N. A impureza adicionada possui 5 elétrons de valência. 

Por exemplo, se um átomo de arsênio for adicionado a um átomo de germânio, quatro dos elétrons de valência se ligarão <br>
aos átomos de Ge, enquanto um elétron permanecerá livre. Isso é ilustrado na figura a seguir. 



Todos esses elétrons livres constituem a corrente eletrônica. Portanto, a impureza, quando adicionada a um semicondutor puro, <br>
fornece elétrons para a condução. 
- Em um semicondutor extrínseco do tipo N, como a condução ocorre por meio de elétrons, os elétrons são os portadores majoritários <br>
e os buracos são os portadores minoritários.
- Como não há adição de cargas positivas ou negativas, os elétrons são eletricamente neutros.
- Quando um campo elétrico é aplicado a um semicondutor do tipo N, ao qual é adicionada uma impureza pentavalente, os elétrons livres <br>
- se movem em direção ao eletrodo positivo. Isso é chamado de condutividade negativa ou do tipo N. 

# Semicondutor extrínseco do tipo P 
Uma pequena quantidade de impureza trivalente é adicionada a um semicondutor puro, resultando em um semicondutor extrínseco <br>
do tipo P. A impureza adicionada possui 3 elétrons de valência. Por exemplo, se um átomo de boro for adicionado a um átomo <br>
de germânio, três dos elétrons de valência se ligam aos átomos de germânio, formando três ligações covalentes. No entanto, <br>
um elétron permanece no germânio sem formar nenhuma ligação. 



A impureza de boro, quando adicionada em pequena quantidade, fornece vários buracos que auxiliam na condução. Todos esses <br>
buracos constituem a corrente de lacunas. 
- Em semicondutores extrínsecos do tipo P, como a condução ocorre através de lacunas, estas são os portadores majoritários, <br>
enquanto os elétrons são os portadores minoritários.
- A impureza adicionada aqui fornece lacunas, que são chamadas de aceptores , porque aceitam elétrons dos átomos de germânio.
- Como o número de lacunas móveis permanece igual ao número de aceptores, o semicondutor do tipo P permanece eletricamente neutro.
- Quando um campo elétrico é aplicado a um semicondutor do tipo P, ao qual é adicionada uma impureza trivalente, os buracos se <br>
movem em direção ao eletrodo negativo, mas a uma velocidade menor do que a dos elétrons. Isso é chamado de condutividade do tipo P.
- Nessa condutividade do tipo P, os elétrons de valência se movem de uma ligação covalente para outra, diferentemente do tipo N.

# Silício
- A diferença de energia entre as bandas é de 0,7 eV, enquanto que para o germânio é de 0,2 eV.
- A geração de pares térmicos é menor.
- A formação da camada de SiO2 é fácil para o silício, o que auxilia na fabricação de muitos <br>
componentes, juntamente com a tecnologia de integração.
- O silício (Si) é encontrado na natureza com mais facilidade do que o germânio (Ge).
- O ruído é menor em componentes feitos de silício (Si) do que em componentes feitos de germânio (Ge).
