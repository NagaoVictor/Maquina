# Lei de Coulomb

A força F entre duas cargas puntiformes q e q' varia diretamente com a grandeza <br>
de cada carga e inversamente com o quadrado da distância r que as separa: <br>
$F = k\frac{qq'}{r^2}$ <br>
onde k é constante de proporcionalidade (dimensional) que depende das unidades <br>
usadas para cargas, distância e força. F será dado em newtons (N) se q e q' forem <br>
em coulombs, r em metros e $k = 9 x 10^9N.m^2/C^2$.

Se definimos $k = \frac{1}{4\pi\epsilon_{0}}$, teremos $F = \frac{1}{4\pi\epsilon_{0}}\frac{qq'}{r^2}$, onde $\epsilon_{0} = \frac{1}{4\pi k} = 8,85 x 10^{-12}C^2/N.m^2$.


# Diferença de Potencial $v$

A diferença de potencial v entre dois pontos é medida de trabalho necessário<br>
à transferência da carga unitária de um ponto para outro. O **volt** é a **diferença** <br>
**de potencial (d.d.p.)** entre dois pontos quando é necessário o trabalho de 1 joule <br>
para a transferência de uma carga de 1 coulomb de um ponto ao outro: 1 joule/coulomb

Uma carga q que se desloca entre dois pontos de um circuito externo, entre os quais <br>
existe uma diferença de potencial v, executa um trabalho medido por $qv$, as se deslocar <br>
do ponto de potencial mais elevado para o mais baixo.

Um elemento ativo, como por exemplo, uma bateria ou um gerador, tem uma **força <br>
eletromotriz (f.e.m)** se ele exerce trabalho sobre uma carga que o atravessa. <br>
A carga recebe energia elétrica ao se deslocar do terminal de menor para o de maior <br>
potencial. A f.e.m é medida pela d.d.p entre os terminais quando o gerador não está <br>
debitando ou drenando corrente.


# Corrente $i$
O material que contém elétrons livres, capazes de se deslocarem de um átomo para o <br>
seguinte, é um condutor. Aplicando-se nele uma diferença de potencial, os elétrons se <br>
deslocam.

Quando uma carga q está sendo transferida de um ponto para outro de um condutor, existe <br>
nele uma corrente elétrica. Se a carga é transferida na razão constante de 1 coulomb/s, <br>
a corrente instantânea $i$ num condutor é:<br>
$i(ampéres) = \frac{dq(coulomb)}{dt(segundos)}$ <br>

O sentido da corrente positiva é, por convenção, oposto àquele em que se deslocam os elétrons.<br>


# Potência $p$
A potência elétrica p é o produto da tensão aplicada $v$ pela corrente resultante $i$.
$p(watt) = v(volts)$ x $i(ampéres)$. <br>

Por definição, a corrente positiva tem a direção da seta na fonte de tensão; ela sai <br>
da fonte pelo terminal +. Quando $p$ é positivo, a fonte transfere energia para o circuito.<br>



Se a potência $p$ é uma função periódica do tempo $t$, de período $T$, a potência média <br>

$P = \frac{1}{T}\int_{0}^{T}p\, dt$ <br>
ou seja: <br>
$P_{média} = \frac{1}{T}\int_{0}^{T}v(t).i(t)\, dt$

# Energia $\omega$
Como a potência é a taxa de transferência da energia em função do tempo, <br>
$p = \frac{d\omega}{dt}$ e $W = \int_{t_1}^{t_2}p\, dt$ <br>
onde W é a energia transferida durante o intervalo de tempo considerado. <br>

# Resistor, Indutor, Capacitor
Ao se fornecer energia elétrica a um elemento de circuito, ele responderá <br>
por uma das seguintes formas:
- A energia é consumida - o elemento de circuito é um resistor puro;
- A energia é armazenada num compo magnético - o elemento é um indutor puro;
- A energia é armazenada num campo elétrico - o elemento é um capacitor puro;

Na prática, um elemento de circuito apresenta mais de uma das características <br>
acima e, talvez, todas as três, simultaneamente. Uma delas pode predominar, entretanto. <br>
Uma bobina pode ser projetada para apresentar elevada indutância, mas o fio com que é <br>
enrolada possui alguma resistência; a bobina apresenta, então, ambas as propriedades.

# Resistência $R$
A diferença de potencial $v(t)$ entre os terminais de um resistor puro é diretamente <br>
proporcional à corrente $i(t)$ que nela circula. A constante de proporcionalidade R <br>
é chamada resistência do resistor e é expressa em volts/amperes ou ohms.<br>
<br>
$v(t) = Ri(t)$ e $i(t)=\frac{v(t)}{R}$ <br>

Nenhuma restrição existe para $v(t)$ e $i(t)$; podem ser constantes em relação ao tempo, <br>
como ocorre nos circuitos de c.c., ou podem ser funções senoidais, co-senoidais etc.

As letras minúsculas $(v,i,p)$ em geral indicam funções do tempo. As maiúsculas $(V,I,P)$ <br>
indicam quantidades contantes; os valores máximos ou **"picos"** recebem um índice <br>
$(V_m, I_m, P_m)$

## Resistores em série:
$R_t = R_1 + R_2 + ... + R_n$
## Resistores em paralélo:
$\frac{1}{R_t} = \frac{1}{R_1} + \frac{1}{R_2} + ... + \frac{1}{R_n}$
## Resistores mistos:
$R_t = (R_1 + R_2 + ... + R_n) + (\frac{1}{R_1} + \frac{1}{R_2} + ... + \frac{1}{R_n}) $

# Indutância $L$

Quando a corrente em um circuito varia, o fluxo magnético que o envolve também varia. <br>
Essa variação de fluxo ocasiona a indução de uma f.e.m.v no circuito. A f.e.m. induzida <br>
$v$ é proporcional à taxa de variação da corrente em relação ao tempo, desde que a <br>
permeabilidade seja constante. A constante de proporcionalidade chamada auto-indutância <br>
ou indutância do circuito.
<br>
$v(t) = L\frac{di}{dt}$ donde $i(t) = \frac{1}{L}\int v \, dt$ <br>

Com $v$ em volts e $di$/$dt$ em amperes/s, $L$ é expressa em volt/ampere ou $henry$. <br>
A auto-indutância de um circuito é de 1 henry (1H) se a f.e.m. nele induzida é de <br>
1 volt, quando a corrente varia à razão de 1 ampere/segundo.

## Indutores em série:
$L_t = L_1 + L_2 + ... + L_n$
## Indutores em paralélo:
$\frac{1}{L_t} = \frac{1}{L_1} + \frac{1}{L_2} + ... + \frac{1}{L_n}$
## Indutores mistos:
$L_t = (L_1 + L_2 + ... + L_n) + (\frac{1}{L_1} + \frac{1}{L_2} + ... + \frac{1}{L_n}) $

# Capacitância $C$
A diferença de potencial $v$ entre os terminais de um capacitor é proporcional <br>
à carga q nele existente. A constante de proporcionalidade $C$ é chamada capacitância <br>
do capacitor. 
<br>
$q(t) = Cv(t)$, $i=\frac{dq}{dt}=C\frac{dv}{dt}$, $v(t)=\frac{1}{C}\int i\, dt$

Com $q$ em coulombs e $v$ em volts, $C$ é obtida em coulombs/volt ou $farads$. <br>
Um capacitor terá a capacitância de 1 farad (1F) se adquirir a carga de 1 coulomb <br>
para cada volt de diferença de potencial aplicada entre seus terminais. São submúltiplos <br>
convenientes do farad:

## Esquemas: 
### Capacitores em Série <br>

$\frac{1}{C_{eq}} = \frac{1}{C_1} + \frac{1}{C_2} \implies C_{eq} = \frac{C_1.C_2}{C_1 + C_2}$<br>

**Carga e Corrente** <br>
$I_{total} = I_1 = I_2 = I_3 = .... = I_n$ <br>
$Q_{total} = Q_1 = Q_2 = Q_3 = .... = Q_n$ <br>
$V_t \cdot C_{\text{eq}} = V_1 \cdot C_1 = V_2 \cdot C_2 = V_3 \cdot C_3 = \dots = V_n \cdot C_n$<br>


**Tensão**<br>
$V_{\text{total}} = V_1 + V_2 + V_3 + \dots + V_n$

Encontre a tensão e corrente e capacitância geral, e em cada pedaço use apenas uma das duas! <br>
A corrente não decai no circuito em série, por isso, use-a em todos os pedaços para saber a queda de tensão $V_{a}-V_{b} = \frac{Q}{C}$


# Leis de Kirchhoff
A soma das correntes que chegam a um nó é igual à soma das correntes que dele saem.<br>
Se as correntes que se dirigem para um nó são consideradas positivas, e negativas as<br>
que dele se afastam, a lei estabelece que é nula a soma algébrica de todas as correntes <br>
que concorrem em um mesmo nó.

A soma das elevações de potencial ao longo de qualquer circuito fechado é igual à soma <br>
das quedas de potencial nesse circuito. Em outras palavras, a soma algébrica das diferença <br>
de potencial, ao longo de um circuito fechado, é nula. Se existir mais de uma fonte e os sentidos <br>
não forem iguais, será considerada positiva a tensão da fonte cujo sentido coincidir com o admitido <br>
para a corrente.
<br>
$\sum$corrente entrando = $\sum$corrente saindo <br>
$i_1 + i_3 = i_2 + i_4 + i_5$ <br>
$i_1 + i_3 - i_2 - i_4 - i_5 = 0$

<br>
$\sum$elevações de potencial = $\sum$quedas de potencial <br>
$v_a - v_b = Ri + L(\frac{di}{dt})$ <br>
$v_a - v_b - Ri - L(\frac{di}{dt}) = 0$

| Capacitância C | $v(t) = \frac{1}{C}\int i\, dt$ | $i(t) = C\frac{dv}{dt}$ |

# Valores Médio e Eficaz
## Formas de Ondas

As representações gráficas de $v(t), i(t), p(t)$ etc. são, respectivamente, as formas de ondas<br>
da tensão, da corrente e da potência. A introdução à análise dos circuitos encara, apenas, as <br>
funções periódicas, isto é, aquelas para as quais $f(t) = f(t+nT)$, onde n é um inteiro e T é <br>
o período. Se a função for periódica, necessário se torna, pelo menos, a representação de um <br>
período para que se possa falar em forma de onda.

As funções tensão e corrente, $v(t)$ e $i(t)$, são expressões matemáticas que podem apresentar-se <br>
sob diversos aspectos. Por exemplo, as funções seno e co-seno podem ser expressas por séries infinitas <br>
de potências. Deve-se ressaltar que as equações básicas que relacionam tensão e corrente, para os três <br>
elementos de circuito, aplicam-se independentemente da forma matemática.

<br>

# Valor Médio
A função periódica geral $y(t)$, de período $T$, tem para valor médio <br>
$$
Y_{med} = \frac{1}{T}\int_{0}^{T} y(t)\, dt
$$

# Valor Médio Quadrático Eficaz ou Efetivo
Da corrente $i(t)$ em um resistor puro $R$ resulta uma potência $p(t)$ de valor médio $P$. <br>
Esta mesma potência $P$ poderia ser produzida em $R$ por uma corrente constante $I$. Diz-se, <br>
então, que a corrente $i(t)$ tem um valor efetivo ou eficaz $I_{ef}$ (também se usa $rms$ <br>
abrev. de $\color{green}{root}$ $\color{green}{mean}$ $\color{green}{square}$, em vez de $ef$), equivalente a essa corrente constante $I$. <br>
O mesmo se aplica a função tensão, cujo valor eficaz é $V_{ef}$. A função geral $y(t)$, de <br>
período $T$, tem um valor eficaz $Y_{ef}$  dada por
$$
Y_{ef} = \sqrt{\frac{1}{T} \int_{0}^{T} y(t)^2 \, dt}
$$

# Valor Eficaz ou Efetivo de Vários Termos Senoidais e Co-Senoidais

A função $y(t) = a_0 + (a_1\cos{\omega t} + a_2\cos{2\omega t} + ...) + (b_1\cos{\omega t} + b_2\cos{2\omega t} + ...)$ <br>
tem valor eficaz dado por <br>
$Y_{rms} = \sqrt{a_0^2 + (A_1^2 + A_2^2 + ...) + (B_1^2 + B_2^2 + ...)}$ <br>
Se for valor eficaz de $a_1\cos{\omega t}$, então $A_1 = \frac{a_1}{\sqrt{2}}$, $A_1^2 = \frac{a_1^2}{2}$ e <br>
$Y_{rms} = \sqrt{a_0^2 + \frac{1}{2}(a_1^2 + a_2^2 + ...) + \frac{1}{2}(b_1^2 + b_2^2 + ...)}$

# Fator e Forma
A relação entre o valor eficaz e o valor médio é o fator de forma $F$ da <br>
onda. Ele é útil na geração de tensão e nos fatores de correção dos instrumentos. <br>
Fator de Forma = $\frac{Y_{ef}}{Y_{med}} = \frac{\sqrt{\frac{1}{T}\int_{0}^{T}y(t)^2 \, dt}}{\frac{1}{T}\int_{0}^{T}y(t)\, dt}$

Outras formas de ondas, como mostra na figura abaixo, podem ter valor médio nulo, sem possuírem <br>

# Corrente e Tensão Senoidais

Quando se aplicam as leis de **$\color{blue}{Kirchhoff}$** a um circuito, resulta, <br>
geralmente, uma equação íntegro-diferencial. Os métodos clássicos de resolução de <br>
equações diferenciais darão as soluções. Quando tais métodos são usados na obtenção <br>
da corrente devida a uma tensão aplicada, a corrente fica dividida em duas partes. <br>
Uma parte transiente, isto é, passageira, transitória, durando, geralmente, apenas <br>
uma fração de segundo e uma parte correnspondente ao regime estacionário, que permanece <br>
até que outra causa seja introduzida.



# Correntes Senoidais
Quando a corrente é senoidal em elementos puros R, L e C, a tensão nos terminais de casa <br>
um é data por:

**Tensão no Elemento Puro para Corrente Senoidal**:
|Elemento|Tensão para $i$ geral| Tensão para $i=I_m\sin\omega t$|Tensão para $i=I_m\cos\omega t$|
|---|---|---|---|
|Resitência R| $ V_R = Ri $ | $ V_R = RI_m\sin\omega t $ | $ V_R = RI_m\cos\omega t $ |
|Indutância L| $ v_L = L\frac{di}{dt} $ | $ V_L=\omega L I_m  \cos\omega t$ | $ v_L = \omega L I_m(-\sin \omega t) $ |
|Capacitância C| $ V_c = \frac{1}{C}\int i\, dt $ | $ v_C = \frac{I_m}{\omega C }(-\cos \omega t) $ | $ v_C = \frac{I_m}{\omega C} \sin\omega t $ |

**Tensão no Elemento Puro para Tensão Senoidal**:
|Elemento|Corrente para $v$ geral| Correte para $v=V_m\sin\omega t$|Corrente para $v=V_m\cos\omega t$|
|---|---|---|---|
|Resitência R| $ i_R = \frac{v}{R} $ | $ i_R = \frac{V_m}{R}\sin\omega t $ | $ i_R = \frac{V_m}{R}\cos\omega t $ |
|Indutância L| $ i_L = \frac{1}{L}\int v\, dt $ | $ i_L=\frac{V_m}{\omega L}   (-\cos\omega t)$ | $ i_L = \frac{V_m}{\omega L} \sin\omega t $ |
|Capacitância C| $ i_C = C\frac{dv}{dt} $ | $ i_C = \omega CV_m\cos \omega t $ | $ i_C = \omega CV_m (-\sin \omega t) $ |

# Impedância

A impedância de um elemento, de um ramo ou de um circuito completo é a relação entre a <br>
tensão e a corrente:
$Impedância = \frac{FunçãoTensão}{FunçãoCorrente}$<br>
Para tensão e corrente senoidais esta relação terá uma amplitude (módule) e um ângulo. <br>

# Ângulo de Fase
Se tensão e corrente forem, ambas, funções senoidais do tempo, a representação gráfica de ambas, <br>
sobre a mesma escala de tempo, mostrará um deslocamento entre elas, salvo se for o caso de uma <br>
resistência pura. Esse deslocamento é o ângulo de fase e nunca excede 90$\celsius$ ou $\pi /2$ radianos. Por <br>
convenção, o ângulo de fase é sempre "o ângulo que a corrente $i$ faz com a tensão $v$", isto é, $i$ está <br>
avançada de 90$\celsius$, em relação à $v$, num capacitor puro; $i$ está atrasada de 45$\celsius$ em relação à $v$, num <br>
circuito $RL$ série em que $R$ é igual a $\omega L$; $i$ está em fase $v$ numa resistência pura etc. O resumo a <br>
seguir esclarece a impedância e o ângulo de fase.


# Elemento R
A corrente e a tensão estão em fase, num resistor puro. <br>
 <br>

# Elemento L
Em um indutor puro, a corrente fica atrasada de 90$\celsius$ ou $\pi /2$ em relação à tensão. <br>
 <br>
O módulo da impedância é $\omega L$.

# Elemento C
No capacitor puro a corrente fica adiantada de 90$\celsius$ ou $\pi /2$ em relação à tensão <br>
 <br>
O módulo da impedância é $\frac{1}{\omega C}$

# RL em série
Corrente atrasada de $\arctan(\frac{\omega L}{R})$ em relação à tensão. <br>
 <br>
O módulo da impedância é $\sqrt{R^2 + (\omega L)^2}$

# RC em série
Corrente avançada de $\arctan(\frac{1}{\omega CR})$ em relação à tensão em um circuito RC em série.<br>
 <br>
O módulo de impedância é $\sqrt{R^2 + (\frac{1}{\omega CR})^2}$

# Números Complex

## Números Reais
O conjunto dos números reais consta dos números racionais e dos irracionais. O conjunto de todos os números reais <br>
pode ser representado, numa correspondência um a um, pelo conjunto de todos os pontos de uma linha reta, chamada <br>
linha (eixo) dos números reais, de modo que cada ponto represnte um único número de linha.

<br>
As operações de adição, subtração, multiplicação e divisão podem ser efetuadas com quaisquer números desse conjunto. <br>
Raízes reais de números positivos podem ser representadas na linha dos números reais, mas a raiz quadrada de um número <br>
negativo não existe no conjunto dos números reais.

## Números Imaginários
A raiz quadrada de um número real negativo é chamada um número imaginário puro. <br>
Ex: $\sqrt{-1},\sqrt{-2},\sqrt{-5},\sqrt{-16}.$ <br>

Se fizermos $j = \sqrt{-1}$, vem $\sqrt{-2} = j\sqrt{2}$, $j\sqrt{-4} = j2$, $j\sqrt{-5}=j\sqrt{5}$ etc. <br>
Segue-se, também, que:<br>
$j^0 = 1$<br>
$j^1 = j$<br>
$j^2 = -1$<br>
$j^3 = -j$<br>

Todos os números imaginários puros podem ser reprenstados por pontos de uma linha reta chamada *linha* (eixo) <br>
dos números imaginários.
<br>

A escolha da palavra imaginário foi inadequada, pois os números imaginários existem, da mesma maneira que os <br>
reais. O termo significa, simplesmente, que tais números não podem ser representados no eixo dos números <br>
reais; estão situados em uma segunda linha, o eixo dos números imaginários.

# Números Complexos
Um número complexo **Z** é um número da forma $x+jy$, onde x e y são reais e j = $\sqrt{-1}$. Em um número <br>
complexo $x+jy$, o primeiro termo, $x$, é chamado parte real e o segundo, $y$, parte imaginária. Quando $x=0$, <br>
o número complexo reduz-se a imaginário puro e corresponde a um ponto do eixo $j$. Do mesmo modo, se $y=0$, o <br>
número complexo é um número real e corresponde a um ponto do eixo real. Os números complexos, portanto, compreendem <br>
todos os números reais e todos os números imaginários puros.

Dois números complexos, $a+jb$ e $c+jd$, serão iguais, e somente o serão, se $a=c$ e $b=d$.

Como na figura a seguir, o eixo dos números reais for perpendicular ao dos números imaginários (ou eixo dos $j$) <br> 
no ponto de cruzamento 0, cada ponto do plano complexo resultante representa um único número complexo e reciprocamente. <br>
$Z_1 = 6$<br>
$Z_2 = 2-j3$<br>
$Z_3 = j4$<br>
$Z_4 = -3+j2$<br>
$Z_5 = -4-j4$<br>
$Z_6 = 3+j3$<br>




# Outras Formas de Números Complexos

 <br>
Na figura acima, $x = \cos \theta$, $y = r\sin\theta$ e o número complexo **Z** é <br>
$Z = x+jy = r(\cos\theta + j\sin\theta)$ onde $r = \sqrt{x^2 + y^2}$ é chamado <br>
módulo ou valor absoluto de *Z* e o ângulo $\theta = \arctan \frac{y}{x}$ é chamado <br>

A fórmula de ***Euler***, $e^{j\theta} = {\cos\theta + j\sin\theta}$, possibilita outra <br>
forma para os números complexos, chamadas forma exponencial.
$Z = r\cos\theta + jr\sin\theta = re^{j\theta}$

A forma polar ou Steinmetz para um número complexo **Z** é bastante usada em análise de <br>
circuitos e escreve-se $z = r \angle \theta$ onde $\theta$ aparece, geralmente, em graus. 

Esses quatro meios de se representar um número complexo estão resumidos a seguir. O emprego<br>
de um ou outro depende da operação a ser efetuada.
- Forma retangula                   $Z = x+jy$
- Forma polar ou de Steinmetz       $Z = r \angle \theta$
- Forma exponencial                 $Z = re^{j\theta}$
- Forma trigonométrica              $Z = r(\cos\theta + j\sin\theta)$

# Conjugado de um Número Complexo
O conjugado **Z*** de um número complexo **Z** $ = x+jy$ é o número complexo **Z*** = $x-jy$. <br>
Por exemplo, $3-j2$ e $3+j2$, $-5+j4$ e $-5-j4$ são dois pares de números complexo conjugados.

Na forma polar, o conjugado de **Z** = $r \angle \theta = r \angle -\theta$. O conjugado de <br>
**Z** = $r(\cos \theta + j\sin \theta)$ é **Z*** = $r(\cos \theta - j\sin \theta)$, pois <br>
$\cos (-\theta)$ = $\cos (\theta)$ e $\sin(-\theta)=-\sin\theta$. Por exemplo, o conjugado <br>
de **Z** = $7\angle 30\degree$ é $7\angle -30\degree$

O conjugado **Z*** de um núemro complexo **Z** é sempre a imagem de **Z** em relação ao eixo <br>
real.

Podemos, então, resumir as quatro formas de se escrever o número complexo **Z** e seu conjugado.
- **Z** = $x+jy$
- **Z*** = $x-jy$
- **Z** = $re^{j\theta}$
- **Z*** = $re^{-j\theta}$
- **Z** = $r \angle \theta$
- **Z*** = $r \angle -\theta$
- **Z** = $r(\cos\theta + j\sin \theta)$
- **Z*** = $r(\cos\theta - j\sin \theta)$
- $z_1 = 3 + j4, z_1$*$ = 3 - j4$
- $z_2 = 5\angle{143,1\degree}, z_2$*$ =5\angle{-143,1\degree}$



# Soma e Diferença de Números Complexos
Para somar ou subtrair dois números complexos somam-se ou subtraem-se separadamente as partes<br>
reais e as parte imaginárias. Do ponto de vista prático, a soma e a subtração de números complexos <br>
só podem ser efetuadas, convenientemente, quando ambos estão na forma retangular.

Exemplo 1 Dados:
$Z_1 = 5 - j2$ e $Z_2 = -3 -j8$. Então, <br>
$Z_1 + Z_2 = (5-3) + j(-2-8) = 2-j10$ <br>
$Z_2 - Z_1 = (-3-5) + j(-8+2) = -8-j6$

# Multiplicação de Números Complexos
O produto de dois números complexos, estandp ambos na forma exponencial, deduz-se diretamente das leis <br>
dos expoentes:
$Z_1Z_2 = (r_1e^{j\theta_1})(r_2e^{j\theta_2})=r_1r_2+2e^{j(\theta_1 + \theta_2)}$ <br>
O produto na forma polar ou de Steinmetz segue-se da forma exponencial: <br>
$Z_1Z_2 = (r_1\angle\theta_1)(r_2\angle\theta_2) = r_1r_1\angle{\theta_1 + \theta_2} $ <br>
O porduto pode ser obtido na forma retangular, tratando-se os dois números complexos <br>
como se fossem binômios.

$Z_1Z_2 = (x_1 + jy_1)(x_2 + jy_2) = x_1x_2 + jx_1x_2 + jy_zx_2 + j^2y_1y_2$ = <br>
$(x_1x_2 - y_1y_2) + j(x_1y_2 + y_1x_2)$

Exemplo 1: <br>
$Z_1 = 5e^{j\frac{\pi}{3}}$ e $Z_2 = 2e^{-j\frac{\pi}{6}}$, então $Z_1Z_2 = (5e^{j\frac{\pi}{3}})(2e^{-j\frac{\pi}{6}}) = 10e^{j\frac{\pi}{6}}$ <br>
Exemplo 2: <br>
$Z_1 = 2\angle 30\degree$ e $Z_2 = 5\angle -45\degree$, então $Z_1Z_2 = (2\angle 30\degree)(5\angle -45\degree) = 10\angle -15\degree$ <br>
Exemplo 3: <br>
$Z_1 = 2+j3$ e $Z_2 = -1-j3$, então $Z_1Z_2 = (2+j3)(-1-j3) = 7-j9$.

# Divisão de Números Complexos
O quociente de dois números complexos na forma exponencial deduz-se diretamente da lei <br>
dos expoentes <br>
$\frac{Z_1}{Z_2} = \frac{r_1e^{j\theta_1}}{r_2e^{j\theta_2}} = \frac{r_1}{r_2}e^{j(\theta_1 - \theta_2)}$

Na forma polar ou de Steinmetz, com referência à forma exponencial:<br>
$\frac{Z_1}{Z_2} = \frac{r_1\angle\theta_1}{r_2\angle\theta_2} = \frac{r_1}{r_2}\angle{\theta_1-\theta_2}$

A divisão na forma retangular se faz multiplicando-se numerador e denominador pelo conjugado do denominador:<br>
$\frac{Z_1}{Z_2} = \frac{x_1 + jy_1}{x_2 + jy_2}(\frac{x_2 + jy_2}{x_2 + jy_2}) = \frac{(x_1x_2 + y_1y_2) + j(y_1x_2 - y_2x_1)}{x_2^2 + y_2^2}$

Exemplo 4:<br>
$Z_1 = 4e^{j\frac{\pi}{3}}$ e $Z_2 = 2e^{j\frac{\pi}{6}}$, vem $\frac{z_1}{z_2} = \frac{4e^{j\frac{\pi}{3}}}{2e^{j\frac{\pi}{6}}} = 2e^{j\frac{\pi}{6}}$

Exemplo 5:<br>
$Z_1 = 8\angle-30\degree$ e $Z_2 = 2\angle-60\degree$, vem $\frac{8\angle-30\degree}{2\angle-60\degree} = 4\angle30\degree$

Exemplo 6:<br>
$Z_1 = 4-j5$ e $Z_2 = 1+j2$, vem $\frac{Z_1}{Z_2} = \frac{4-j5}{1+j2}(\frac{1-j2}{1-j2}) = \frac{-6-j13}{5}$

# Raízes de Números Complexos
Qualquer número complexo **Z** = $re^{j\theta}$ pode ser escrito **Z** = $re^{j(\theta + 2\pi n)}$, <br>
onde $ n = 0,\pm 1, \pm 2, ...$ Da mesma maneira **Z** = $r\angle\theta$ pode ser escrito **Z** = $r\angle(\theta + n360\degree)$. <br>
Então, <br> 
**Z** = $re^{j\theta} = re^{j(\theta + 2\pi n)} $ e $\sqrt[k]{Z} = \sqrt[k]{r} e^{\frac{j(\theta + 2\pi)}{k}}$ <br>

**Z** = $r\angle \theta = r\angle (\theta n360\degree)$ e $\sqrt[k]{Z} = \sqrt[k]{r}\angle\frac{\theta + n360\degree}{k}$ <br>

Portanto, as k raízes distintas do número complexo podem ser obtidas atribuindo-se a $n$ os valores $0, 1, 2, 3, ...k-1$

Exemplo 7:<br>
Se **Z** $= 8\angle 60\degree$, tem-se $\sqrt[3]{Z} = \sqrt[3]{8}\angle \frac{(60\degree + n360\degree)}{3} = 2\angle (20\degree + n120\degree)$. <br>
Fazendo-se $n$ iguais a 0,1 e 2, obtêm-se as três raízes cúbicas, que são: $2\angle 20\degree$,$2\angle 140\degree$ e $2\angle 260\degree$. <br>

Exemplo 8:<br>
Achar as cinco raízes quintas da unidade. <br>
Como $1 = 1e^{j2\pi n}$, vem: $\sqrt[5]{1} = \sqrt[5]{1}e^{\frac{j2\pi n}{5}} = 1e^{\frac{j2\pi n}{5}}$. <br>
Atribuindo-se a $n$ os valores 0, 1, 2, 3 e 4, obtêm-se as cinco raízes <br>
$1\angle 0\degree$ ou $1$; $1\angle 72\degree$; $1\angle 144\degree$; $1\angle 216\degree$ e $1\angle 288\degree $

# Logaritmo de um Número Complexo
O logaritmo neperiano de um número complexo pode ser determinado com facilidade, a partir de sua forma exponencial.

O resultado não é único. Usa-se, mais frequentemente, o valor principal, quando $n=0$. <br>
Exemplo 9: <br>
**Z** $= 3e^{\frac{j\pi}{6}}$, tem-se $\ln 3 e^{\frac{j\pi}{6}} = \ln3 + \frac{j\pi}{6} = 1,099+j0,523$.

# Forma Retangular para Forma Polar
Exemplo 12: <br>
Exprimir $4 + j3$ em forma polar, $r\angle \theta$.
1) Faça um esboço exagerando o fato de que a parte real é maior do que a parte imaginária, <br>
quer dizer, o ângulo é menor do que $45\degree$.
2) $\theta = \arctan\frac{3}{4}=\arctan0,75=36,9\degree$ $(<45\degree)$ <br>
$\sin 36,9\degree = 0,600$, donde $r=\frac{3}{0.600} = 5$ ou $\cos36,9\degree = 0,800$ sendo, <br>
pois, $r=\frac{4}{0,800} = 5$.
3) $4+j3=5\angle 36,9\degree$

<br>
