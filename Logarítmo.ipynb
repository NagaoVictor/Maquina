# Criação Logarítmica
Para $b,a > 0$ <br>
Se $b^x = a$, então $\log_{b}{a}$<br>
Então $\log_{b}{a} = \mathbb{R}$ <br>

# Estrutura Logarítmica
$\log a = \log_{10}{a}$ <br>
$\log_{b}{a^x} = x\log_{b}{a}$ <br>
$\log_{b^x}{a} = \frac{1}{x}\log_{b}{a}$ <br>
$\log_{b}{a} = \frac{\log_{c}{a}}{\log_{c}{b}}$ <br>
$\log_{b}{a} = \frac{1}{\log_{a}{b}}$ <br>
$\log_{c}{a.b} = \log_{c}{a} + \log_{c}{b}$ <br>
$\log_{c}{\frac{a}{b}} = \log_{c}{a} - \log_{c}{b}$ <br>
$\log_{b}{a} = x \implies b^x = a$


# Comportamento Logarítmico

Para $b,a < 0 \to \log_{b}{a} = $ Número Interito Ímpar <br>

Para $0 < b < a < 1$:


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(2, 15, 100)
y = 2/x

def log_simples(i, j):
    return np.log10(2/i)/np.log10(2/j**2)

def log_simples2(i, j):
    return np.log10(2/j**2)/np.log10(2/i)

def log_sqrt(i):
    return np.sqrt(i)

fig, ax = plt.subplots(3,3)

ax[0,0].plot(x, log_simples(x,y))
ax[0,1].plot(x, log_simples2(x,y))
```

```python
log_simples(x,y) > log_simples2(x,y)
```
