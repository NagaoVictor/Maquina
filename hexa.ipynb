# Cálculos de Hexadecimal
Para facilitar os cálculos e compreender como funciona dígitos hexadecimais

$Hexa = \sum_{i=0}^n{K_i.16^i}$
- Hexa = Modelo Hexadecimal
- {K_i} = Dígito Constante Hexadecimal
- i = Index do Dígito

# Modo de calcular
Use o método hexa com um parâmetro.
- Digite uma string com os bits do registrador.

# Calculo de Hexadecimal para Decimal
Uso o método calculo para descobrir o decimal.
- Digite uma string com os bits do registrador.
```python
calculo("1131")
```

**retorno**
```income 
**retornando**
4371
```


````python
## Hexadecimal

def hexa(digito:str) -> str:
    barramento = len(digito)
    decimal_word = ""
    for i in reversed(range(0, barramento)):
        decimal_word += f"{digito[barramento-(i+1)]}.16^{i} "
    return decimal_word
````

````python
hexa("1113")
````

````python
def calculo(digito:str) -> int:
    barramento = len(digito)
    valor_total = 0
    for i in reversed(range(0, barramento)):
        valor_total += int(digito[barramento-(i+1)])*(16**i)
    return valor_total
````

````python
calculo("1113")
````
