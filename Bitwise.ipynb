# Operanções Básicas BitWise 

## AND (&)

```python
a = 0b1010
b = 0b1100

resultado = a & b
print(bin(resultado))  # Saída: 0b1000
```

## OR (|)

```python
a = 0b1010
b = 0b1100

resultado = a | b
print(bin(resultado))
```

## XOR (^)

```python
a = 0b1010
b = 0b1100

resultado = a ^ b
print(bin(resultado))  # Saída: 0b0110
```

## NOT (~)

```python
a = 0b1010

resultado = ~a
# Nota: Em Python, inteiros têm tamanho dinâmico (infinitos bits na teoria),
# então o NOT exibe o complemento de dois com sinal. Para ver limpo em 4 bits, usamos uma máscara:
print(bin(resultado & 0b1111))  # Saída: 0b0101 (inverteu 1010 para 0101)
```

## Shift Left (<<)

```python
a = 0b0001  # Decimal 1

resultado = a << 2  # Desloca 2 posições para a esquerda
print(bin(resultado))  # Saída: 0b0100 (Decimal 4)
```

## Shift Right (>>)

```python
a = 0b1000  # Decimal 8

resultado = a >> 2  # Desloca 2 posições para a direita
print(bin(resultado))  # Saída: 0b0010 (Decimal 2)
```

```python
# Simulando um registrador de 32 bits em Python com um valor inicial qualquer
registrador = 0b11110000111100001111000011110000

print("Registrador original: ", bin(registrador))

# Definimos o que queremos alterar
novo_valor = 0b1010  # O valor 10 em binário
posicao_inicial = 4  # O campo começa no bit 4
tamanho_campo = 4  # O campo tem 4 bits de largura (do 4 ao 7)

# 1. Criar a máscara dinamicamente
# (1 << tamanho_campo) - 1 cria uma sequência de '1's do tamanho do campo (ex: 4 bits = 0b1111)
# Depois deslocamos para a posição correta
mascara = ((1 << tamanho_campo) - 1) << posicao_inicial
print("Máscara gerada:       ", bin(mascara))

# 2. Limpar o campo (zera apenas os bits daquele campo usando AND com o inverso '~')
registrador_limpo = registrador & ~mascara
print("Registrador limpo:    ", bin(registrador_limpo))

# 3. Inserir o novo valor na posição correta (usando OR)
registrador_atualizado = registrador_limpo | (novo_valor << posicao_inicial)
print("Registrador final:    ", bin(registrador_atualizado))
```

```python
valor = 0b10110101
print(valor.bit_count())  # Saída: 5 (tem cinco bits 1)
```

```python
# Convertendo um número inteiro para uma sequência de 2 bytes (Big-Endian)
dado_16bits = 0x1234
bytes_prontos = dado_16bits.to_bytes(2, byteorder="big")
print(bytes_prontos)  # Saída: b'\x124' (dois bytes prontos para o I2C)

# O caminho inverso (reconstruir o inteiro a partir dos bytes)
original = int.from_bytes(bytes_prontos, byteorder="big")
print(hex(original))  # Saída: 0x1234
```
