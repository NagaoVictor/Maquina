# Tabela Periódica
## Familias
Família 1 (Exceto o Hidrogênio): Metais Alcalinos
- Metais macios, altamente reativos (especialmente ccom a àgua) e possuem apenas 1 elétron na camada de valência
- Li(Lítio), Na(Sódio), K(Potássio), Rb(Rubídio), Cs(Césio), Fr(Frâncio)

Família 2: Metais Alcalino-Terrosos
- Menos reativos que os metais alcalinos, possuem 2 elétrons na camada de valência e são encontrados abundantemente <br>
na crosta terrestre na forma de compostos.
- Be(Berílio), Mg(Magnésio), Ca(Cálcio), Sr(Estrôncio), Ba(Bário), Ra(Rádio)

Família 3 à 12: Metais de transição
- Apresentam alta dureza, elevados pontos de fusão e ebulição, além de serem excelentes condutores de eletricidade e calor. <br>
Muitos formam compostos coloridos.
- Sc(Escândio), Ti(Titânio), V(Vanádio), Cr(Cromo), Mn(Manganês), Fe(Ferro), Co(Cobalto), Ni(Níquel), Cu(Cóbre), Zn(Zinco), <br>
Y(Ítrio), Zr(Zircônio), Nb(Nióbio), Mo(Molibdênio), Tc(Tecnécio), Ru(Rutênio), Rh(Ródio), Pd(Paládio), Ag(Prata), Cd(Cádmio), <br>
Hf(Háfnio), Ta(Tântalo), W(Tungstênio), Re(Rênio), Os(Ósmio), Ir(Íridio), Pt(Prata), Au(Ouro), Hg(Mercúrio), <br>
Rf(Rutherfórdio), Db(Dúbnio), Sg(Seabórgio), Bh(Bóhrio), Hs(Hássio), Mt(Meitnério), Ds(Darmstádtio), Rg(Roentgênio), Cn(Copernício)

Família 13: Família do Boro (Icosagênios)
- Contém um semimetal (boro) e metais com 3 elétrons na camada de valência. O alumínio é o metal mais abundante na crosta <br>
terrestre.
- B(Boro), Al(Alumínio), Ga(Gálio), In(Índio), Tl(Tálio), Nh(Nihônio)

Família 14: Família do Carbono (Tetragênios)
- Possuem 4 elétrons na camada de valência. Incluem não-metais (carbono), semimetais (silício e germânio) e metais. <br>
São fundamentais para a química orgânica e a indústria de semicondutores.
- C(Carbono), Si(Silício), Ge(Germânio), Sn(Estanho), Pb(Chumbo), Fl(Fleróvio)

Família 15: Família do Nitrogênio (Pnicogênios)
- O nome "pnicogênio" deriva do grego pnigein (sufocar). Possuem 5 elétrons na camada de valência e são essenciais para <br>
a fertilidade do solo e a vida biológica.
- N(Nitrogênio), P(Fósforo), As(Arsênio), Sb(Antimônio), Bi(Bismuto), Mc(Moscóvio)

Família 16: Calcogênios
- O nome significa "geradores de cobre" (ou criadores de minérios). Possuem 6 elétrons na camada de valência; o oxigênio é <br>
o elemento mais abundante no corpo humano e na crosta terrestre.
- O(Oxigênio), S(Enxofre), Se(Selênio), Te(Telúrio), Po(Polônio), Lv(Livermório)

Família 17: Halogênios
- O nome significa "geradores de sais". Possuem 7 elétrons na camada de valência, são altamente reativos, tóxicos em estado <br>
puro e formam sais iônicos com os metais alcalinos.
- F(Flúor), Cl(Cloro), Br(Bromo), I(Iodo), At(Ástato), Ts(Tenesso)

Família 18: Gases Nobres
- Caracterizam-se pela estabilidade química e baixa reatividade, pois possuem a camada de valência completa (geralmente 8 elétrons, <br>
exceto o hélio com 2). Encontram-se no estado gasoso em condições ambiente.
- He(Hélio), Ne(Neônio), Ar(Argônio), Kr(Criptônio), Xe(Xenônio), Rn(Radônio), Og(Oganessônio)

## Metais de Transição Interna (Séries Inferiores)
Lantanídeos
- Elementos de número atômico de 57 a 71, conhecidos historicamente como "Terras raras".
- La(Lantânio), Ce(Cério), Pr(Praseodímio), Nd(Neodímio), Pm(Promécio), Sm(Samário), Eu(Európio), Gd(Gadolínio), Tb(Térbio) <br>
Dy(Disprósio), Ho(Hólmio), Er(Érbio), Tm(Túlio), Yb(Itérbio), Lu(Lutécio)

Actinídeos
- Elementos de número atômico de 89 a 103, que incluem todos os elementos radiotivos.
- Th(Tório), Pa(Protactínio), U(Urânio), Np(Neptúnio), Pu(Plutônio), Am(Amerício), Cm(Cúrio), Bk(Berquélio), Cf(Califórnio) <br>
Es(Einstênio), Fm(Férmio), Md(Mendelévio), No(Nobélio), Lr(Laurêncio)

# Estrutura sintática

${}^{A}_{z}\text{E}^{+ ou -}$
- z = Prótons (Número atômico)
- n = Neutrons 
- A = p + n (Massa Atômica)
