```python
import random as rd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
capital = 100
bet_size = 1
prob_win = 0.51
prob_loss = (1 - prob_win)
win_reward = 1
loss_reward = -1
```

```python
fig, ax = plt.subplots(figsize=(16,8))
profit = []
for j in range(100):
    equity_curve = [capital]
    for i in range(1000):
        if prob_loss < rd.random():
            equity_curve += [equity_curve[-1] + (win_reward*bet_size)]
        else:
            equity_curve += [equity_curve[-1] + (loss_reward*bet_size)]
    profit += [equity_curve[-1] - capital]
    ax.plot(equity_curve)
```

```python
import numpy as np
mean_profit = np.mean(profit)
print(f'$ {mean_profit} ')
```

```python
capital = 100
bet_size = 10
bet_size_start = 10
prob_win = 0.50
prob_loss = (1 - prob_win)
win_reward = 1
loss_reward = -1
```

```python
fig, ax = plt.subplots(figsize=(16,8))
for j in range(100):
    bet_size = 10
    equity_curve = [capital]
    for i in range(1000):
        if prob_loss < rd.random():
            equity_curve += [equity_curve[-1] + (win_reward*bet_size)]
            bet_size = bet_size_start
        else:
            equity_curve += [equity_curve[-1] + (loss_reward*bet_size)]
            bet_size = bet_size * 2
        if equity_curve[-1] <= -500:
            break
        elif equity_curve[-1] >= 2000:
            break
    ax.plot(equity_curve)
```
