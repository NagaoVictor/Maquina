# **Motores de Combustão Interna**

As máquinas térmicas são dispositivos que permitem transformar calor em trabalho. <br>
O calor pdoe ser obtido de diferente fontes: combustão, energia elétrica, energia atômica etc. <br>
Este texto preocupa-se apenas com o caso em que o calor é obtido pela queima do combustível, <br>
isto é, energia química em trabalho mecânico. <br>

### **Balanceamento** 

$Gas_{0} + Combustível \to Gas_{-1} + Trabalho + Perda De Calor$


A obtenção de trabalho é ocasionada por uma sequência de processos realizado numa substância que <br>
será denominada "fluido ativo - FA". No caso do balanceamento, o FA é formado pela mistura ar e combustivel <br>
na entrada do volume de controle e produtos da combustão na saída. <br>

Quanto ao comportamento do fluido ativo (FA), as máquinas térmicas serão classificadas em:
    Motores de combustão externa - ***MCE***: quando a combustão se processa externamente ao FA,<br>
    que será apenas o veículo da energia térmica a ser transformada em trabalho, como, por exemplo uma <br>
    máquina a vapor usando um ciclo de Rankine.

### **Ciclo Rankine**

#### **Compressão Isentrópica (A Bomba)**
O que acontece: O fluido (água no estado líquido) entra em uma bomba a baixa pressão e é comprimido até alta <br>
pressão necessária para entrar na caldeira.

O fluido da bomba precisa entrar na caldeira que está à elevados níveis de pressão. Para ocupar esses sistema, é necessário <br>
empurrar com uma bomba para ganho de pressão no líquido, e entrar direto na caldeira.

$Energia_{líquido_{0}} + Trabalho \to Energia_{líquido_{1}}$ <br>
$W = (P_{2}-P_{1})v$ <br>
O trabalho será realizado pela bomba.

#### **Adição de Calor a Pressão Constante (A Caldeira)**
O que acontece: O líquido pressurizado entra na caldeira. Lá, ele absorve uma quantidade massiva de calor, vinda <br>
da queima de combustível ou de fissão nuclear.

O fluido precisa superaquecer para se transformar em gás para alimentar a turbina no próximo passo. <br>
O objetivo da caldeira é transformar calor em temperatura.

$Energia_{interna_{0}} + Trabalho \to Energia_{interna_{1}}$ <br>
$m.c_{e}.T_{1} + W = m.c_{e}.T_{2}$<br>
$m.c_{e}.T_{1} + W = m.c_{e}.(T_{1} + \Delta T)$<br>
$m.c_{e}.T_{1} + W = m.c_{e}.T_{1} + m.c_{e}.\Delta T$<br>
$W = m.c_{e}.\Delta T$<br>


#### **Expansão Isentrópica (A Turbina)**
O que acontece: O vapor de alta pressão passa por uma turbina. À medida que o vapor se expande e perde pressão, ele empurra <br>
as pás da turbina, fazendo o eixo girar em alta velocidade, o que aciona o gerador elétrico.

#### **Rejeição de Calor a Pressão Constante (O Condensador)**
O que acontece: O vapor de baixa pressão (muitas vezes misturado com um pouco de líquido) sai da turbina e entra em um condensador.<br>
Nele, o valpr troca calor com uma fonte fria (como água de um rio ou torre de resfriamento) e se condensa totalmente, virando líquido <br>
saturado novamente.

# Posições do Pistão e Suas Características

$PMS$: Ponto Morto Superior - É a posição na qual o pistão está o mais próximo possível do cabeçote. <br>
$PMI$: Ponto Morto Inferior - É a posição na qual o pistão está o mais afastado possível do cabeçote. <br>
$S$: Curso do pistão - É a distância percorrida pelo pistao quando se desloca de um ponto morto para outro. <br>
$V_{1}$: Volume total - É o volume compreendido entre a cabeça do pistão e o cabeçote, quando o pistão está no PMI <br>
$V_{2}$: Volume morto ou volume da câmara de combustão - É o volume compreendido entre a cabeça do pistão e cabeçote, <br>
quando o pistão está no PMS (também indicado com $V_{m}$) <br>
$V_{du}$: Cilindrada unitária - Também conhecida como volume deslocado útil ou deslocamento volumétrico, é o volume deslocado <br>
pelo pistão de um ponto morto a outro. <br>
$z$: Número de cilindros do motor. <br>
$D$: Diâmetro dos cilindros do motor. <br>
$V_{d}$: Volume deslocado do motor, deslocamento volumétrico do motor ou cilindrada total.<br>
$R_{v}$: Relação volumétrica ou taxa de compressão - É a relação entre o volume total ($V_{1}$) e o volume morto ($V_{2}$), e representa <br>
em quantas vezes $V_{1}$ é reduzido.


# Equações
#### Cilindradas Unitárias

$V_{du} = V_{1} - V_{2}$ <br>
$V_{du} = \frac{\pi D^2 }{4}S$

#### Volume Deslocado

$V_{d} = V_{du}.z = \frac{\pi D^2 }{4}S.z$

#### Taxa de compressão

$R_{v} = \frac{V_{1}}{V_{2}}$ onde $V_{1} \geq V_{2}$ <br>
$V_{du}+V_{2} = V_{1}$ <br>
$R_{v} = \frac{V_{du}+V_{2}}{V_{2}}$ = $\frac{V_{du}}{V_{2}} + 1$


````python
import numpy as np

class Dinamica:
    def __init__(self, curso_pistao, diametro, taxa = 0.035 ,quantidade_cilindros = 4): # (cm, cm)
        self.curso_pistao = curso_pistao # Tamanho do Cilindro sem o incremento
        self.incremento_cilindro_volume = (np.pi*(diametro**2)*curso_pistao*taxa) # Porcentagem já calculando o volume
        self.diametro = diametro
        self.quantidade_cilindros = quantidade_cilindros
    
    def volume_total(self): # Volume da parte cilíndrica mais para da câmara em centimetros cúbicos
        return (np.pi*(self.diametro**2)*self.curso_pistao/4) + self.incremento_cilindro_volume
        
    def cilindrada_unitaria(self): # Volume de cada parte cilindrica de cada pistão
        return np.pi*(self.diametro**2)*self.curso_pistao/4 

    def volume_deslocado(self): # Volume total de deslocamento de todos os pistãos
        return self.cilindrada_unitaria()*self.quantidade_cilindros

    def taxa_compressao(self): # Porcentagem de compressão 
        return (self.cilindrada_unitaria() /self.incremento_cilindro_volume) + 1

d = Dinamica(50, 8)
d.cilindrada_unitaria()
````

### Dinâmica do Motor

É exigido quatro parâmetros para calcular a dinâmica do motor.
- Curso do pistão
- Diâmetro do pistão
- Taxa: Valor da câmara quando o pistão está em PMS
- Quantidade de cilindros

O curso do pistão e o diâmetro, é basicamente, medir o tamanho da parte cilíndrica e a base de uma ponta à outra respecticamente,em centimetros.
A taxa, é relacionada diretamente ao valor do volume do cilindro. Como esta parte é irregular, é necessário o uso de modelagens <br>
matematicas para criar um equação para cada modelo específico. Para não precisar mudar o foco, será usado 3,5% do volume total do <br>
cilindro. Se quiser colocar o valor préviamente calculado, basta seguir as instruções abaixo:

# Nomenclatura Cinemática

$V.E.$: Válvula de escapamento <br>
$V.A.$: Válvula de admissão <br>
$r$: Raio da manivela <br>
$n$: Frequência da árvore de manivelas <br>
$\omega$: Velocidade angular da árvore de manivelas <br>
$V_{p}$: Velocidade média do pistão <br>
$S = 2.r$<br>
$\omega = 2\pi.n$<br>
$V_{p} = 2.S.n$ <br>
