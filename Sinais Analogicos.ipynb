# Onda
Uma onda é uma perturbação física ou eletromagnética que propaga energia através de um meio ou do vácuo sem transportar matéria. <br>
Ela é o conceito unificador entre a física dos sólidos, a teoria de circuitos e o processamento avançado de sinais.

**Parâmetros Físicos Fundamentais**
- Amplitude (A): A altura máxima da oscilação em relação ao ponto de equilíbrio. Em termos práticos, dita a potência, o volume de um <br>
som ou a intensidade de um sinal elétrico.
- Frequência ($f$): O número de ciclos completos que ocorrem em um segundo, medida em Hertz (Hz). Define o tom de um som ou a rapidez <br>
de um sinal de rádio.
- Período ($T$): O intervalo de tempo exato necessário para a onda completar um ciclo inteiro. Relaciona-se inversamente com a frequência <br>
através da fórmula $T = 1/f$.
- Comprimento de Onda ($\lambda$): A distância física real entre dois pontos equivalentes consecutivos (como de uma crista à outra), medida <br>
em metros.
- Velocidade de Propagação ($v$): A rapidez com que a frente de onda viaja pelo espaço, vinculada pela relação $v = \lambda \cdot f$.
- Fase ($\phi$): O deslocamento angular ou temporal da onda em relação a um referencial zero, indicando onde o ciclo começou.

**Classificação das Ondas**
**Natureza:**
- Mecânicas: Precisam de um meio material para se propagar (exemplos: ondas sonoras no ar, ondas na água).
- Eletromagnéticas: Propagam-se no vácuo e através de campos elétricos e magnéticos oscilantes (exemplos: <br>
luz visível, micro-ondas, sinais de Wi-Fi e rádio AM/FM).

**Direção de Vibração:**
- Transversais: A vibração ocorre de forma perpendicular à direção de propagação da onda (ex: luz, ondas em cordas).
- Longitudinais: A vibração ocorre na mesma direção da propagação, gerando compressões e rarefações (ex: som).

# Moldes 

**O Molde Polinomial (Distorção Harmônica)**
- Matemática: $y(t) = \alpha_1 x(t) + \alpha_2 x^2(t) + \alpha_3 x^3(t)$
- Função: Insere potências do sinal linear. O termo quadrático ($x^2$) gera deslocamento de frequência e componentes<br>
de segunda harmônica, enquanto o cúbico ($x^3$) introduz intermodulação e expansão/compressão de ganho.

**O Molde de Saturação (Clipping / Limitador Físico)**
- Matemática: $y(t) = \max(-L, \min(L, x(t)))$
- Função: Trava o sinal linear em um limite máximo $L$ e mínimo $-L$. Simula o comportamento real de amplificadores <br>
operacionais ou transistores operando na região de corte/saturação.

**O Molde de Multiplicação Cruzada (Mixer / Modulador)**
- Matemática: $y(t) = x_1(t) \cdot x_2(t)$
- Função: Multiplica dois sinais lineares independentes. Como a multiplicação de funções não é uma operação linear <br>
(viola o princípio da superposição), ela cria novas frequências (soma e diferença), base essencial para modulações como <br>
AM e processamento de áudio (efeito ring modulator).

**O Molde de Retificação de Meia Onda (Função Não-Linear de Limiar)**
- Matemática: $y(t) = \max(0, x(t))$
- Função: Descarta completamente a porção negativa de um sinal linear alternado (como um diodo ideal), transformando uma <br>
onda senoidal simétrica em uma sequência de pulsos unidirecionais.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-30, 30, 1000)
angle = np.linspace(-90, 90, 1000)

def sinusoidal_signal(A, w0,phi):
    return A*np.cos(w0 + phi)

def sinc_function(t):
    return np.sin(t*np.pi)/(t*np.pi) 

def sampling_function(t):
    return np.sin(t)/t

#plt.plot(angle, sinusoidal_signal(10, 3, angle))
plt.plot(x, sampling_function(x))
plt.plot(x, sinc_function(x))
```

```python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6,3))

x = np.linspace(0, 30, 1000)

def senoid_frequency(x, A, f):
    return A*(np.cos(2*np.pi*f*x))

def senoid_period(x, A, T):
    return A*(np.cos(2*np.pi*x/T))

ax.plot(x, senoid_frequency(x, 10, 0.2))
ax.plot(x, senoid_frequency(x, 10, 0.1))
ax.plot(x, senoid_period(x, 10, 5))
ax.plot(x, senoid_period(x, 10, 4))
```

# Expressões matemáticas 

**Sinal Modulador m(t)**: <br>
$m(t) = A_m \cos(2\pi f_m t)$
- $A_m$: Amplitude do sinal modulador
- $f_m$: Frequência do sinal modulador

**Sinal Portador c(t)**: <br>
$c(t) = A_c \cos(2\pi f_c t)$
- $A_c$: Amplitude do sinal portador
- $f_c$: Frequência do sinal portador

**Sinal Resultante s(t)**: <br>
$s(t) = [A_c + A_m \cos(2\pi f_m t)]\cos(2\pi f_c t)$

# Reconstituição do Formato do Sinal Resultante
O objetivo da modulação AM tradicional é fazer com que a amplitude da onda portadora varie diretamente <br>
em proporção ao sinal da mensagem $m(t)$.
- Passo 1 (Deslocamento DC): Somando a amplitude da portadora $(A_c)$ ao sinal da mensagem $m(t)$. A amplitude <br>
resultante em cada instante passa a ser:
$A(t) = A_c + m(t)$
- Passo 2 (Substituição): Substituímos o sinal mensagem $m(t)$ pela sua forma matemática real: $A(t) = A_c + A_m \cos(2\pi f_m t)$
- Passo 3 (Construção do Sinal Modulado): Multiplicamos essa nova amplitude modulada pela onda portadora de alta frequência original <br>
$s(t) = [A_c + A_m \cos(2\pi f_m t)]\cos(2\pi f_c t)$
