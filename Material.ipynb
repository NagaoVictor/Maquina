## 1. Propriedades Mecânicas
- **Dureza:** Resistência à deformação plástica localizada (risco ou indentação).
- **Módulo de Elasticidade (Módulo de Young - E):** Medida da rigidez elástica e da força de coesão atômica.
- **Limite de Escoamento (σ_y):** Tensão limite que marca a transição do regime elástico para o plástico (deformação permanente).
- **Resistência à Tração (UTS):** Tensão máxima suportada pelo material antes do início da estrição e ruptura.
- **Tenacidade:** Capacidade de absorver energia e deformar plasticamente antes de fraturar.
- **Ductilidade / Fragilidade:** Grau de deformação plástica suportado até a ruptura (alto vs. baixo).
- **Resiliência:** Capacidade de absorver energia estritamente no regime elástico.

## 2. Propriedades Térmicas
- **Condutividade Térmica (k):** Taxa de condução de calor por elétrons livres (metais) ou fônons (cerâmicas/polímeros).
- **Calor Específico (c_p):** Energia térmica necessária para elevar a temperatura de uma unidade de massa em 1 °C.
- **Coeficiente de Expansão Térmica (α):** Taxa de variação dimensional em função da temperatura (governada pela anarmonicidade atômica).
- **Temperatura de Fusão (T_m) e Transição Vítrea (T_g):** Pontos de transição de fase (cristalina para líquida ou amorfa rígida para flexível).

## 3. Propriedades Magnéticas
- **Permeabilidade (μ) e Suscetibilidade (χ):** Facilidade de magnetização do material.
- **Diamagnetismo:** Resposta fraca e repulsiva (elétrons emparelhados).
- **Paramagnetismo:** Atração fraca devido a dipolos desalinhados.
- **Ferromagnetismo:** Alinhamento espontâneo de spins com forte retenção de magnetização (ex: Ferro, Níquel, Cobalto).
- **Antiferromagnetismo e Ferrimagnetismo:** Alinhamentos complexos com cancelamento parcial ou total de spins.
- **Coercividade (H_c) e Remanência (B_r):** Parâmetros de histerese para ímãs permanentes ou materiais eletromagnéticos moles.

## 4. Propriedades Elétricas e Eletrônicas
- **Condutividade (σ) e Resistividade (ρ):** Facilidade ou oposição ao fluxo de corrente elétrica.
- **Classificação de Banda:**
  - **Condutores:** Bandas superpostas com elétrons livres abundantes.
  - **Isolantes:** Bandgap largo que impede a excitação térmica de portadores.
  - **Semicondutores:** Bandgap estreito com condutividade controlável.
- **Supercondutividade:** Resistência elétrica nula abaixo de uma temperatura crítica (T_c) por pares de Cooper.
- **Termoeletricidade:** Conversão direta entre gradientes de temperatura e tensão elétrica (Efeitos Seebeck e Peltier).

## 5. Propriedades Ópticas
- **Índice de Refração (n):** Razão de velocidade da luz que dita o desvio angular (refração).
- **Absortividade, Reflectividade e Transmitividade:** Coeficientes de passagem, reflexão e retenção de fótons.
- **Luminescência (Fluorescência / Fosforescência):** Emissão de luz decorrente de excitação eletrônica.

## 6. Propriedades Químicas e Eletroquímicas
- **Resistência à Corrosão e Oxidação:** Estabilidade contra degradação ambiental por reações eletroquímicas.
- **Potencial de Eletrodo:** Tendência do material a atuar como ânodo ou cátodo em pilhas galvânicas.
- **Catalisidade:** Capacidade de acelerar reações químicas superficiais sem sofrer consumo intrínseco.
