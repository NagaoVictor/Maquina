# Intrdução

A ***$\color{blue}{termodinâmica}$*** é a ciência que estuda o armazenamento, a transformação e a transferência de **energia**. <br>
A energia é armazenada como **energia interna** (Associada com a temperatura), **energia cinética** (devido ao <br>
movimento), **energia potencial** (devido à elevação) e **energia química** (devido à composição química); essa <br>
energia é transfomada de um desses tipos apra o outro e é transferida através de uma fronteira na forma de **calor** <br>
ou **trabalho**. Na *termodinâmica*, produzimos **equações matemáticas** que relacionam as transformações e transferências <br>
de energia com propriedades dos materiais, como **temperatura**, **pressão** e **entalpia**. Assim, as substânias e suas propriedades <br>
passam a ser tema secundário importante. Boa parte desse trabalho se baseará na ***primeira e segunda lei da termodinâmica*** em sistemas <br>
macroscópicos. <br>



# Sistemas termodinâmicos e volume de controle

## Sistemas Termodinâmicos
Um Sistema é definido estritamente como uma quantidade fixa e identificável de matéria escolhida para estudo. <br>
**Abordage Lagrangeana**: Se a matéria se move, expande ou comprime, o sistema vai junto com ela. <br>
**Fronteira do Sistema**: (Linha real ou imaginária) que separa o sistema de sua vizinhança. Ela pode se mover <br>
ou mudar de formato mas nenhuma massa cruza essa fronteira. A massa total permanece constante ($m=constante$).

## Classificação dos Sistemas Termodinâmicos
Os sistemas fechados (massa de controle) são classificados com base no tipo de interação que suas fronteiras permitem <br>
com a vizinhança.
**Sistema Fechado Simples**:
- Permite apenas troca de energia (na forma de calor e trabalho) com o exterior.
- Proibido troca de massa.
- Exemplo: Gás confinado dentro de um cilindro fechado por um pistão.

**Sistema Isolado**:
- É um caso limite onde a fronteira é totalmente impermeável tanto para massa quanto para energia (calor ou trabalho).
- *Regra*: $\Delta{E_{sistema} = 0}$ e $m = constante$.
- Exemplo: Um modelo ideal de garrafa térmica perfeitamente vedada e rígida.



## Volume de controle
Se o Sistema é a abordagem da "partícula" ou da "massa que viaja", o Volume de Controle é a abordagem da "estação de monitoramento".

**Definição Fundamental** (Abordagem Euleriana)
No sistema, a fronteira se move com a matéria, no **Volume de Controle** é uma região fixa no espaço (Ou velocidade constante) através <br>
da qual a matéria flui.
- Fronteira: **Chamada de Superfície de Controle (SC)**. Ela é estática em relação a um referencial escolhido.
- Massa: A massa não é constante dentro do volume. Ela entra e sai continuamente.

A aplicação de volume de controle, é normalmente usada quando o sistema possui algum tipo de escoamento.


# Descrição Macroscópica
**Postulado**: Para trabalhar com a termodinâmica, o material no sistema ou no volume de controle é contínuo. <br>
Ou seja, ele é distribuido continuamente por toda região de interesse. Isso permite o uso de poucas propriedades <br>
mensuráveis.

***Densidade*** <br>
$\rho = \lim_{\Delta{V} \to 0} \frac{\Delta{m}}{\Delta{V}}$

O $\Delta{m}$ é a massa contida no volume $\Delta{V}$. Fisicamente, $\Delta{V}$ não pode diminuir até zero, isso fere <br>
a característica contínua, além de que, para existir massa, precisa existir um objeto. Para isso, é usado uma constante <br>
denominada $\epsilon$ que representa uma quantidade mesmo que mínima de cerca de $3.10^{16}$ para o volume mínimo.


# Propriedades e Estados de um Sistema
A matéria de um sistema pode existir em diversos estados: sólidos, líquidos ou gasosos. Uma fase é uma <br>
quantidade de matéria que possui a mesma composição química em todo seu conjunto, ou seja, que é homogênea. <br>
As fronteira de fase separa as fases de um todo, chamado de *mistura*.

**Propriedades vs Interações**<br>
Na termodinâmica, uma propriedade é qualquer característica ou quantidade mensurável que serve para descrever o estado de um sistema em um determinado instante.
- Pressão
- Volume
- Temperatura
- Energia Interna

**Propriedades (Pressão, Temperatura, Volume, Energia Interna)**: São funções de estado. O valor acumulado em um ciclo fechado é sempre zero ($\oint d\phi = 0$).

A Característica Essencial de uma Propriedade Para que uma quantidade seja considerada uma propriedade legítima na termodinâmica, ela precisa obedecer a uma regra de ouro: ela não tem memória do passado. <br>
- **Valor Único por Estado**: Em um dado estado de equilíbrio, a propriedade tem um valor fixo e exclusivo. <br>
- **Independência da Trajetória (Função de Estado)**: O valor de uma propriedade não depende de como o sistema chegou àquele estado. Se você mudar o <br>sistema de um Estado 1 para um Estado 2, a variação da propriedade dependerá exclusivamente dos extremos, nunca do caminho que você fez:$$\Delta \phi = \phi_2 - \phi_1$$

Enquanto as propriedades descrevem o que o sistema tem (armazenado estaticamente em um estado), as interações descrevem o que o sistema faz ou recebe enquanto muda de estado.
- Calor
- Trabalho

**Propriedades Extensivas e Intensivas** <br>
Relações de tamano e quantidade.

**Propriedades Extensivas** <br>
São aquelas cujos valores dependem diretamente da extensão, do tamanho ou da quantidade de massa presente no sistema. <br>
Se for dividido o sistema ao meio, o valor de uma propriedade extensiva se divide proporcionalmente. Se você juntar dois <br>
sistemas idênticos, os valores dessas propriedade se somam. <br>
Exemplo:
- Massa (m): Medida da quantidade de matéria.
- Volume (v): Espaço ocupado pelo sistema.
- Energia Interna Total (U): Entalpia (H), Entropia Total (S): Quantidade absolutas de energia ou desordem armazenadas.
