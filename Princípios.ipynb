# Princípio da Casa de Pombos (Pigeonhole Principle)
- O Conceito: Se $n$ itens são colocados em $m$ recipientes, e $n > m$, pelo menos um recipiente conterá mais de um item.
- Aplicação na Computação: É a base para provar a existência de ciclos em sequências finitas (como no Algoritmo de Ciclo de Floyd), colisões em funções de hash e a periodicidade de restos em aritmética modular (essencial para entender o comportamento cíclico das potências e a função totiente de Euler).

# Princípio da Invariância (Invariant Principle)
- O Conceito: Uma propriedade que permanece verdadeira antes, durante e depois de cada iteração de um algoritmo ou transformação de estado.
- Aplicação na Computação: É a ponte direta entre a matemática pura e o código. Usamos invariantes validados por asserções (assert) para garantir que o estado de um loop (como nas somas telescópicas ou no Algoritmo de Euclides) mantenha a correção matemática até a terminação.

# Princípio da Simetria e Fatores Complementares
- O Conceito: Se um número inteiro $n$ possui um divisor $d$, então $n = d \cdot c$, o que implica obrigatoriamente que um dos fatores ($d$ ou $c$) deve ser menor ou igual a $\sqrt{n}$.
- Aplicação na Computação: Fundamenta a otimização de algoritmos de primalidade e fatoração. Em vez de testar divisores até $n$ (tempo linear $O(n)$), reduzimos o espaço de busca estritamente até a raiz quadrada, atingindo eficiência em $O(\sqrt{n})$.

# Princípio do Menor Contra-Exemplo (Redução ao Absurdo via WOP)
- O Conceito: Para provar que uma proposição é verdadeira para todos os naturais, assumimos por absurdo que existe um conjunto não-vazio de exceções (contra-exemplos). Pelo Princípio da Boa Ordenação, esse conjunto deve possuir um menor contra-exemplo. Ao analisar esse caso mínimo, demonstramos uma contradição lógica, provando que nenhuma exceção pode existir.
- Aplicação na Computação: Usado formalmente na verificação de correção de algoritmos e na prova de unicidade de estruturas (como a unicidade da fatoração no Teorema Fundamental da Aritmética).

# Princípio da Conservação de Estrutura (Homomorfismos e Congruências)
- O Conceito: Propriedades aritméticas (como soma e multiplicação) podem ser preservadas mesmo quando mapeadas para um espaço restrito de restos (anéis de resíduos $\mathbb{Z}_m$).
- Aplicação na Computação: Permite realizar operações com números gigantescos na criptografia (como o RSA) sem sofrer estouro de memória ou de bits de hardware, operando sempre dentro de limites controlados através de reduções modulares aplicadas em cada passo intermediário.

# Princípio da Descida Estrita (Algorithmic Descent)
- O Conceito: Garante que se você gerar uma sequência de inteiros não-negativos estritamente decrescente a cada passo ($r_0 > r_1 > r_2 > \dots \ge 0$), essa sequência tem um fim obrigatório em tempo finito, fundamentado diretamente no Princípio da Boa Ordenação.
- Aplicação na Computação: É a garantia matemática exata de que o Algoritmo de Euclides para o cálculo do MDC nunca entrará em loop infinito. A cada iteração, o resto da divisão torna-se estritamente menor que o divisor anterior, convergindo necessariamente para zero.

# Princípio da Geração de Ideais (Principal Ideal Domain Property)
- O Conceito: No conjunto dos números inteiros ($\mathbb{Z}$), qualquer conjunto de combinações lineares da forma $ax + by$ (onde $x, y \in \mathbb{Z}$) forma um ideal principal. O menor inteiro positivo pertencente a esse conjunto é exatamente o $\text{MDC}(a, b)$.
- Aplicação na Computação: Explica o funcionamento do Algoritmo de Euclides Estendido. Ele não apenas acha o MDC, mas revela os coeficientes exatos $x$ e $y$ da Identidade de Bézout ($ax + by = \text{MDC}(a, b)$), permitindo calcular inversos modulares para a criptografia.

# Princípio da Conservação de Restos (Modular Substitution)
- O Conceito: Se um número $a$ pode ser escrito como $bq + r$, então o Máximo Divisor Comum entre $a$ e $b$ é exatamente igual ao Máximo Divisor Comum entre $b$ e o resto $r$ ($\text{MDC}(a, b) = \text{MDC}(b, r)$).
- Aplicação na Computação: Permite substituir números gigantescos por restos pequenos em tempo de execução. Em vez de operar com $a$ e $b$ massivos, reduzimos o problema iterativamente mantendo a mesma integridade matemática com uma fração microscópica do custo computacional.

# Princípio da Combinação Linear (Identidade de Bézout)
- O Conceito: O conjunto de todos os números que podem ser formados por combinações lineares de $a$ e $b$ (ou seja, $ax + by$ para quaisquer inteiros $x, y$) forma exatamente um conjunto de múltiplos inteiros do $\text{MDC}(a, b)$.
- Aplicação na Computação: É o princípio que valida o Algoritmo de Euclides Estendido. Ele prova que sempre é possível encontrar coeficientes $x$ e $y$ tais que $ax + by = \text{MDC}(a, b)$. Se $\text{MDC}(a, b) = 1$, conseguimos isolar o fator que gera o inverso modular, essencial para a criptografia moderna.

# Princípio do Homomorfismo Modular (Redução Antecipada)
- O Conceito: As operações aritméticas básicas (soma, subtração e multiplicação) "comutam" com a operação de resto. Em termos práticos: calcular o resto de uma operação gigantesca no final é matematicamente idêntico a aplicar o operador módulo (%) em cada etapa intermediária.
- Aplicação na Computação: Evita completamente o estouro de memória (overflow) em linguagens como C ou Python quando lidamos com números massivos. Aplicar (a * b) % m passo a passo mantém os valores dentro de limites seguros de hardware sem perder a precisão do resultado final.

# Princípio da Inversão Modular (Coprimalidade e Resolução Única)
- O Conceito: A divisão convencional não existe diretamente na aritmética modular (você não pode simplesmente "dividir por $b$" em anéis de resíduos se o resto não for exato). No entanto, se $\text{MDC}(a, m) = 1$, existe garantidamente um inverso multiplicativo único $a^{-1}$ tal que $(a \cdot a^{-1}) \equiv 1 \pmod m$.
- Aplicação na Computação: Permite substituir operações de divisão inteira problemáticas por uma multiplicação pelo inverso modular, o que abre portas para resolver sistemas de congruências lineares de forma ultra-eficiente.

# Propriedades do Cálculo Numérico

1. mod (Módulo / Operação Resto)
- O que é: É a operação matemática que calcula o resto da divisão inteira entre dois números.
- Exemplo: $17 \text{ mod } 5 = 2$ (porque $17$ dividido por $5$ dá $3$, e o que sobra — o resto — é $2$).
- Na prática: É a "aritmética do relógio". Quando o relógio chega em 12, ele zera e volta para o 1 (modo 12). <br>
Na criptografia,o módulo limita os números dentro de um ciclo fixo.



2. Congruência Modular ($\equiv$)
- O que é: É o equivalente ao "igual" (=) na matemática comum, mas adaptado para o mundo dos restos.
- Significado: Dizemos que dois números são congruentes se eles deixam o mesmo resto quando divididos pelo mesmo módulo.
- Exemplo: $15 \equiv 1 \pmod 7$. (Tanto o 15 quanto o 1 deixam resto $1$ quando divididos por $7$: $15 \div 7$ sobra $2$ com resto $1$; e $1 \div 7$ sobra $0$ com resto $1$).

3. MDC (Máximo Divisor Comum / GCD)
- O que é: O maior número inteiro positivo que consegue dividir dois números ao mesmo tempo, sem deixar resto.
- Exemplo: $\text{MDC}(12, 18) = 6$ (o maior número que divide tanto o 12 quanto o 18 é o 6).
- Na prática: Na criptografia, o MDC serve como um "teste de segurança". Se o MDC entre dois números for maior que 1, <br>
significa que eles compartilham divisores, o que quebra a segurança do sistema (por isso precisamos que o MDC seja estritamente 1).

4. Coprimos (ou Primos entre si)
- O que é: Dizemos que dois números $A$ e $B$ são coprimos quando o $\text{MDC}(A, B) = 1$. Eles não precisam <br>
ser números primos individualmente; o importante é que eles não compartilham nenhum divisor comum além do 1.
- Exemplo: $8$ e $9$ são coprimos? O 8 divide por {1, 2, 4, 8} e o 9 divide por {1, 3, 9}. O único divisor comum <br>
é o $1$. Logo, $\text{MDC}(8, 9) = 1$ (eles são coprimos, mesmo nenhum dos dois sendo primo).

5. Totiente de Euler ($\phi(N)$)
- O que é uma função totiente: É uma função matemática criada por Leonhard Euler que conta quantos números inteiros <br>
positivos menores do que $N$ são coprimos com $N$.
- O atalho mágico: Se $N$ for o produto de dois números primos ($p$ e $q$), calcular o totiente vira uma conta simples:
  $$\phi(N) = (p - 1) \times (q - 1)$$
- Na prática: É o tamanho do "espaço de segurança" que usamos para calcular as chaves do RSA.

6. Inverso Modular
- O que é: É o equivalente à divisão ou ao número recíproco no mundo modular. O inverso modular de um número $E$ <br>
é um número $D$ tal que, quando você os multiplica, o resultado dá 1 em relação ao módulo.
- Fórmula: $E \times D \equiv 1 \pmod{\text{Módulo}}$
- Na prática: É a nossa Chave Privada. Se a chave pública é $E$, a chave privada é o seu inverso modular $D$, <br>
calculado usando o Algoritmo de Euclides Estendido.

7. Exponenciação Modular
- O que é: A operação de elevar um número a uma grande potência dentro de um ciclo de módulo ($Base^{Expoente} \pmod{Módulo}$).
- Na prática: É o motor de força bruta controlada da criptografia. É a conta pesada que transforma a mensagem na cifra <br>
- ($m^e \pmod N$) e a cifra na mensagem ($c^d \pmod N$).

8. Máximo Divisor Comum (MDC)
- Propriedade da Redução Euclidiana: $\text{MDC}(a, b) = \text{MDC}(b, a \text{ mod } b)$. É a base recursiva que <br>
transforma um problema gigantesco em divisões sucessivas rápidas.
- Divisibilidade Comum: Qualquer divisor comum de $a$ e $b$ também divide obrigatoriamente o $\text{MDC}(a, b)$.

9. Identidade de Bézout e Algoritmo Estendido
- Propriedade da Combinação Linear: Se $d = \text{MDC}(a, b)$, então existem sempre inteiros $x$ e $y$ tais que $ax + by = d$.
- Garantia de Solutividade: Uma equação diofantina linear $ax + by = c$ só tem solução inteira se $c$ for um múltiplo exato de <br>
-  $\text{MDC}(a, b)$.

10. Aritmética Modular e Congruências
- Compatibilidade com Adição e Multiplicação: Se $a \equiv b \pmod n$ e $c \equiv d \pmod n$, então:
    - $(a + c) \equiv (b + d) \pmod n$
    - $(a \cdot c) \equiv (b \cdot d) \pmod n$

- Cancelamento Restrito: Só podemos cancelar um fator comum $c$ em uma congruência ($ac \equiv bc \pmod n$) <br>
se $\text{MDC}(c, n) = 1$.

11. Inverso Modular
- Condição de Existência: O inverso modular de $a$ módulo $n$ só existe se e somente se $\text{MDC}(a, n) = 1$
(ou seja, se forem coprimos).
- Unicidade no Anel: O inverso modular é único dentro do intervalo de $0$ a $n-1$.

12. Teoremas de Fermat e Euler
- Pequeno Teorema de Fermat: Se $p$ é primo e $a$ não é divisível por $p$, então $a^{p-1} \equiv 1 \pmod p$.
- Teorema de Euler (Generalização): Se $\text{MDC}(a, n) = 1$, então $a^{\phi(n)} \equiv 1 \pmod n$. É o passaporte <br>
matemático que garante que o RSA funciona ao resetar os ciclos do expoente.

13. Primos, Crivos e Primalidade
- Teorema Fundamental da Aritmética: Todo número inteiro maior que 1 é um número primo ou é um produto único de números <br>
primos (desconsiderando a ordem).
- Congruências de Fermat para Testes: Se um número $n$ é primo, ele satisfaz $a^{n-1} \equiv 1 \pmod n$ para qualquer base<br>
$a$ (embora o inverso nem sempre seja verdade, base dos testes probabilísticos).

14. Teorema Chinês do Resto (TCR)
- Propriedade de Isomorfismo: Um sistema de congruências simultâneas com módulos coprimos entre si possui uma solução <br>
única módulo o produto de todos os módulos ($N = m_1 \cdot m_2 \cdots m_k$).

15. Ordens Modulares e Estrutura do RSA
- Teorema de Lagrange em Grupos Modulares: A ordem (o tamanho do ciclo) de qualquer elemento divide estritamente <br>
o tamanho do grupo multiplicativo ($\phi(n)$).
- Propriedade da Função de Mão Única: Elevar à potência pública é computacionalmente trivial ($O(\log e)$), mas <br>
extrair a raiz módulo $N$ sem a chave privada é intratável.

16. Exponenciação Modular Rápida (Square-and-Multiply)
- Propriedade da Decomposição Binária: Qualquer expoente $b$ pode ser escrito como uma soma de potências de 2. <br>
Isso reduz a complexidade de calcular $a^b \pmod n$ de $O(b)$ (força bruta linear) para $O(\log b)$ (tempo logarítmico).

17. Polinômios em Corpos Finitos
- Propriedade da Divisão Algébrica de Polinômios: Assim como nos inteiros, dados dois polinômios $A(x)$ e $B(x)$, <br>
existem polinômios únicos $Q(x)$ (quociente) e $R(x)$ (resto) tais que $A(x) = B(x)Q(x) + R(x)$, onde o grau de <br>
$R(x)$ é menor que o grau de $B(x)$.
- Fechamento em Corpos Finitos ($\mathbb{F}_p$): As operações de adição, subtração e multiplicação de polinômios cujos <br>
coeficientes pertencem a um corpo finito mantêm o resultado dentro do mesmo conjunto restrito por módulo primo.

18. Transformadas Rápidas (FFT e NTT)
- Propriedade da Convolução: A multiplicação de dois polinômios de grau $N$ (que por força bruta custaria $O(N^2)$) <br>
equivale a mapear os pontos, multiplicá-los pontualmente e aplicar a transformada inversa, reduzindo a complexidade <br>
para $O(N \log N)$.
- Raiz da Unidade Discreta: A Transformada Rápida de Fourier baseia-se nas potências de uma raiz primitiva da unidade<br>
($\omega^n = 1$), cujas propriedades de simetria e cancelamento eliminam cálculos redundantes.

19. Métodos de Fatoração de Inteiros
- Propriedade da Congruência de Quadrados: Se conseguirmos encontrar dois números $x$ e $y$ tais que <br>
$x^2 \equiv y^2 \pmod N$ (com $x \not\equiv \pm y \pmod N$), então o $\text{MDC}(x - y, N)$ revela <br>
instantaneamente um fator não trivial de $N$.
- Comportamento Cíclico (Pollard's Rho): Baseia-se no Paradoxo do Aniversário; a função pseudo-aleatória <br>
gera sequências que inevitavelmente entram em colisão cíclica em $O(\sqrt{p})$ passos, permitindo achar <br>
fatores primos ocultos.

20. Representação de Ponto Flutuante (IEEE 754)
- Propriedade da Precisão Finita: Os números reais no computador são representados de forma <br>
discreta pela fórmula $(-1^s) \times 1.m \times 2^{e-bias}$, gerando lacunas intransponíveis <br>
entre representações numéricas consecutivas (definidas pelo Machine Epsilon).
- Perda de Significância por Subtração: Subtrair dois números de ponto flutuante muito próximos <br>
e quase iguais destrói os dígitos significativos mais à esquerda, gerando erros catastróficos <br>
de arredondamento.

21. Erros Computacionais e Propagação
- Propriedade do Erro Absoluto vs. Relativo: O erro absoluto acumula-se linearmente em <br>
operações de soma e subtração, enquanto o erro relativo se propaga de forma multiplicativa <br>
e pode ser amplificado severamente por funções mal condicionadas.
- Instabilidade Numérica: Um algoritmo é considerado instável se pequenos erros iniciais de <br>
arredondamento crescem exponencialmente a cada passo da computação (como em recursências mal <br>
formuladas).

22. Métodos Iterativos (Raízes de Equações)
- Propriedade da Convergência Quadrática (Newton-Raphson): Sob condições de suavidade da função, <br>
o método de Newton dobra o número de casas decimais corretas a cada nova iteração, convergindo <br>
muito mais rápido que o método da bissecção (que tem convergência linear).
- Teorema do Ponto Fixo: Uma função contínua possui pelo menos um ponto fixo ($g(x) = x$) se mapeia <br>
um intervalo fechado nele mesmo, garantindo a convergência de métodos iterativos simples.

23. Integração Numérica e Otimização
- Propriedade da Ordem de Erro nas Quadraturas: Regras de integração baseadas em interpolação <br>
polinomial (como a Regra do Trapézio e a Regra de Simpson) reduzem o erro de discretização em <br>
função de potências negativas do número de subintervalos ($h^2$ ou $h^4$).

24. Aritmética Modular e Estruturas Cíclicas
- Propriedade da Periodicidade dos Restos: As potências sucessivas de qualquer inteiro $a$ <br>
módulo $n$ geram uma sequência que se repete ciclicamente (formando um subgrupo periódico), <br>
cuja extensão máxima é restrita pelo Totiente de Euler $\phi(n)$.
- Propriedade do Resto Negativo: Em operações modulares, $(-a) \pmod n$ é equivalente matemáticamente <br>
a $(n - (a \pmod n))$. Isso evita estouros ou inconsistências lógicas ao lidar com números negativos <br>
em linguagens de programação que tratam o operador % como truncamento em vez de módulo real.

25. Primalidade e Testes Probabilísticos (Miller-Rabin)
- Propriedade das Raízes Não-Triviais da Unidade: Se $n$ é um número primo ímpar e $x$ é um <br>
inteiro tal que $x^2 \equiv 1 \pmod n$, então obrigatoriamente $x \equiv 1 \pmod n$ ou <br>
$x \equiv -1 \pmod n$.
    - Aplicação: Se você achar um $x$ que fure essa regra (uma raiz não-trivial), você provou <br>
    matematicamente que $n$ não é primo (composto). É exatamente assim que o teste de Miller-Rabin <br> 
    caça números compostos em milissegundos.

26. Otimizações do Teorema Chinês do Resto (TCR)
- Propriedade da Forma Canônica de Garner: Em vez de usar a fórmula clássica de somatórias <br>
com inversos globais gigantescos para resolver sistemas de congruências, o algoritmo de Garner <br> 
reconstrui a solução de formaincremental usando coeficientes mistos, evitando estouros de precisão <br>
em inteiros de grande porte.

27. Exponenciação Modular Rápida (Técnicas Avançadas)
- Propriedade da Janela Deslizante (Sliding Window): Em vez de processar o expoente bit a bit <br>
(como no Square-and-Multiply padrão), agrupa-se blocos de bits para pré-computar potências ímpares, <br> 
reduzindo o número total de multiplicações modulares em chaves criptográficas grandes em cerca de  <br>
15% a 25%.

28. Polinômios e Aritmética em Corpos Finitos ($\mathbb{F}_{2^m}$)
- Propriedade da Aritmética Sem Transporte (Carry-less Arithmetic): Em corpos finitos de característica <br>
2 (muito usados em criptografia moderna e criptografia de curvas elípticas), a adição e a subtração de <br>
polinômios são equivalentes à operação lógica XOR ($\oplus$), eliminando completamente o custo de <br>
transporte de bits (carry) do hardware.

29. Convolução e Transformadas (NTT)
- Propriedade do Teorema da Convolução Discreta: A convolução linear de dois vetores de tamanho <br>
$N$ pode ser calculada preenchendo-os com zeros (zero-padding) até o próximo formato potência de <br>
2, aplicando a Transformada Numérica (NTT) de forma direta e evitando loops encaixados de <br>
complexidade quadrática.

30. Cálculo Numérico e Estabilidade (Ponto Flutuante)
- Propriedade da Lei Distributiva Falha: Na aritmética de ponto flutuante IEEE 754, a propriedade <br>
distributiva da matemática ($a \cdot (b + c) = a \cdot b + a \cdot c$) não é estritamente garantida <br>
devido aos truncamentos sucessivos de arredondamento em cada operação isolada.

A Anatomia Matemática (O Limite Formal)Formalmente, dizemos que uma função $f(n)$ é $O(g(n))$ se existirem constantes reais positivas $c$ e $n_0$ tais que:$$0 \le f(n) \le c \cdot g(n), \quad \forall n \ge n_0$$O que isso significa na prática: A partir de um ponto crítico $n_0$, a sua função de custo $f(n)$ é limitável superiormente por um múltiplo constante de $g(n)$.Regra de Ouro da Simplificação: Elimine coeficientes e termos desprezíveis. Se o seu algoritmo executa $3n^2 + 5n + 100$ operações, a assinatura assintótica é $O(n^2)$, pois o termo quadrático domina quando $n$ explode.

* **$O(1)$ — Tempo Constante**
  * **Comportamento Prático ($n = 10^9$):** Instantâneo (independente de $n$).
  * **Algoritmos e Exemplos:** Fórmulas fechadas de Faulhaber, operações de nível de bit (`bitwise` em C/Python), acesso direto a índices em arrays (`arr[i]`) e hashing perfeito.

* **$O(\log \log n)$ — Duplo Logarítmico**
  * **Comportamento Prático ($n = 10^9$):** Praticamente constante (menos de 5 operações).
  * **Algoritmos e Exemplos:** Algoritmo de *Interpolative Search* (Busca Interpolada otimizada) e algoritmos avançados de estruturas de dados sucintas.

* **$O(\log n)$ — Tempo Logarítmico**
  * **Comportamento Prático ($n = 10^9$):** Extremamente rápido ($\approx 30$ operações para $10^9$).
  * **Algoritmos e Exemplos:** **Algoritmo de Euclides** (Cálculo de MDC), **Exponenciação Binária** (*Square-and-Multiply*), busca binária em arrays ordenados e operações em Árvores Binárias de Busca Balanceadas (AVL / Red-Black).

* **$O(\sqrt{n})$ — Tempo Sub-linear / Raiz Quadrada**
  * **Comportamento Prático ($n = 10^9$):** Viável para limites moderados ($n = 10^{12} \implies 10^6$ operações).
  * **Algoritmos e Exemplos:** Fatoração por tentativa de divisão até $\sqrt{n}$, algoritmo *Baby-step Giant-step* para logaritmo discreto e Algoritmo de Ciclo de Floyd (detecção de ciclos).

* **$O(n)$ — Tempo Linear**
  * **Comportamento Prático ($n = 10^9$):** Proporcional à entrada (1 bilhão de passos exatos).
  * **Algoritmos e Exemplos:** **Crivo de Eratóstenes** linear para primos, varredura simples de vetores (*Single Pass*), cálculo de somatórias via laços iterativos simples (`for` loops) e Algoritmo *Kadane* para subarray de soma máxima.

* **$O(n \log \log n)$ — Praticamente Linear**
  * **Comportamento Prático ($n = 10^9$):** Ligeiramente superior a $O(n)$, mas altamente eficiente.
  * **Algoritmos e Exemplos:** Crivo de Eratóstenes otimizado com limites assintóticos finos para geração de primos massivos.

* **$O(n \log n)$ — Linearítmico**
  * **Comportamento Prático ($n = 10^9$):** O limite ótimo para algoritmos baseados em comparação.
  * **Algoritmos e Exemplos:** **Merge Sort** e **Heap Sort**, Algoritmo de *Fast Fourier Transform* (FFT / NTT) para multiplicação rápida de polinômios e Algoritmo de *QuickSort* (no caso médio).

* **$O(n^2)$ — Tempo Quadrático**
  * **Comportamento Prático ($n = 10^9$):** Intratável para $n$ grande ($n = 10^5 \implies 10^{10}$ operações).
  * **Algoritmos e Exemplos:** Busca por força bruta de pares (loops aninhados), algoritmos de ordenação ingênuos (*Bubble Sort*, *Insertion Sort*, *Selection Sort*) e multiplicação clássica de matrizes sem otimização.

* **$O(n^3)$ — Tempo Cúbico**
  * **Comportamento Prático ($n = 10^9$):** Restrito a entradas muito pequenas ($n \le 2500$).
  * **Algoritmos e Exemplos:** **Algoritmo de Floyd-Warshall** para caminhos mínimos entre todos os pares e multiplicação padrão de matrizes ($A \times B$).

* **$O(2^n)$ — Tempo Exponencial**
  * **Comportamento Prático ($n = 10^9$):** Explosão combinatória imediata ($n = 50$ já esgota a máquina).
  * **Algoritmos e Exemplos:** Solução de força bruta para o **Problema da Mochila Binária** (*Knapsack*), geração de todos os subconjuntos (Conjunto das Partes / *Powerset*) e Torres de Hanói resolvidas por recursão pura.

* **$O(n!)$ — Tempo Fatorial**
  * **Comportamento Prático ($n = 10^9$):** Intratável absoluto (inviável mesmo para $n > 20$).
  * **Algoritmos e Exemplos:** Solução de força bruta para o **Problema do Caixeiro Viajante** (TSP) e geração exata de todas as permutações possíveis de um conjunto.

```python
import random as rd

class Algorithm:
    info = "Criação de uma classe que possui uma memória baseada em índice para saber como o algorítmo se desenvolve ao longo de etapas"
    def __init__(self):
        self.counter = 0
        self._lista = []
        self._memoria = {}
        
    def memoria(self):
        if len(self._lista) != 0:
            self._memoria[self.counter]=self._lista
            self.counter+=1
    
    def lista(self, origem=0, intervalo=20, quantidade=15) -> None:
        for i in range(quantidade):
            x = rd.sample(range(origem, intervalo), 1)
            self._lista.append(x[0])
        self.memoria()
        
    def compare(self,anterior, proximo):
        if anterior >= proximo:
            return True
        else:
            return False

    def update(self):
        for i in range(0, len(lista)-1):
            if self.compare(lista[i], lista[i+1]) == True:
                generic = lista[i]
                lista[i] = lista[i+1]
                lista[i+1] = generic
        self.memoria()
```
