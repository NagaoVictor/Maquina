# Eletromagnetismo Introdução
## Equações de Maxwell
$rot \vec{B} = \mu\vec{j} + \mu_0\epsilon_0 \frac{\partial\vec{E}}{\partial t}$ <br>
$rot \vec{E} = -\frac{\partial\vec{B}}{\partial t}$ <br>
$div \vec{E} = \frac{\rho}{\epsilon_0}$ <br>
$div\vec{B}=0$

## Eletrostática (Nada Varia Com o Tempo)
$div \vec{E} = \frac{\rho}{\epsilon_0}$ <br>
$rot \vec{E} = 0$

## Magnetostática (Nada Varia Com o Tempo)
$rot \vec{B} = \mu\vec{j}$ <br>
$rot \vec{B} = 0$

O propósito é trabalhar com as equações de Maxwell, e traduzir para ondas eletromagnéticas propagadas no vácuo.

# Princípio de Superposição
## Lei de Coulomb
 <br>

$F = \frac{k.|Q|.q}{d^2}$ <br>
$k = \frac{1}{4\pi\epsilon_0} = 9\times \frac{10^9 N.m^2}{C^2} $

## Princípio
 <br>

$\vec{F}_R = \vec{F_1} + \vec{F_2} + \vec{F_3}$ <br>

 <br>


**Modulo** (Não vetores) <br>
$F_1 = \frac{K|Q_1||q|}{d_1^2}$ <br>
$F_2 = \frac{K|Q_2||q|}{d_2^2}$ <br>
$F_3 = \frac{K|Q_3||q|}{d_3^2}$ <br>

$F_1 = F_2 = \frac{K|Q||q|}{x^2}$ <br>
$F_R = F_1 \cos{\theta} + F_2 \cos{\theta}$ <br>
$F_R = 2\frac{K|Q||q|}{X^2}\cos{\theta}$ <br>
$F_R = 2\frac{K|Q||q|}{(D^2 + d^2)}\cdot\frac{d}{\sqrt{D^2 + d^2}}$ <br>
$F_R = 2\frac{K|Q||q|d}{(D^2 + d^2)^{\frac{3}{2}}}$



# Princípio de Superposição (Discreto)

 <br>

## Força Entre Duas Cargas
$\vec{F_1} = \frac{1}{4\pi\epsilon_0}\cdot\frac{Q_1q}{|\vec{'r_1}|^2}\hat{'r_1}$ <br>
$\vec{r_1} + \vec{'r_1} = \vec{R}$ <br>
$\vec{'r_1} = \vec{R} - \vec{r_1}$ <br>
**Usando o conceito de Versor**<br>
$\hat{'r_1} = \frac{\vec{'r_1}}{|\vec{'r_1}|}$ 

## Força Entre Multiplas Cargas
$\vec{F_i} = \frac{1}{4\pi\epsilon_0}\cdot\frac{Q_iq}{|\vec{'\vec{r_i}}|^2}\hat{'r_i}$ <br>
$\vec{'r_i} = \vec{R}-\vec{r_i}$ <br>
$\hat{'r_1} = \frac{\vec{'r_1}}{|\vec{'r_1}|}$ <br>

$\vec{F} = \sum_{i=1}^{N}\vec{F_i}=\frac{q}{4\pi\epsilon_0}\cdot\sum_{i=1}^{N}\frac{Q_i}{|\vec{'\vec{r_i}}|^2}\hat{'r_i}$

# Princípio de Superposição (Contínua)

 <br>

## Força Entre Duas Cargas
$\vec{F_1} = \frac{1}{4\pi\epsilon_0}\cdot\frac{Q_1q}{|\vec{'r_1}|^2}\hat{'r_1}$ <br>
$\vec{r_1} + \vec{'r_1} = \vec{R}$ <br>
$\vec{'r_1} = \vec{R} - \vec{r_1}$ <br>
$\rho(\vec{r}) = \frac{\partial Q}{\partial V}$ Volume do cubo <br>
$\vec{F} = \int_{V} d\vec{F}$ <br>
$\vec{F} = \frac{q}{4\pi\epsilon_0}\cdot\int_{V}\frac{\rho(\vec{r})dV }{|\vec{'\vec{r_i}}|^2}\hat{'r_i}$
