# Complex Numbers and 2D Vectors 
$z = a + bj$ <br>

$\operatorname{Re}(z) = \operatorname{Re}(a+bj) = a$ <br>

# Multiplication and Rotations 
$z_1z_2=(a+bj)(c+dj)$ <br>
$z_1z_2=ac+adj+bcj+bdj^2$ <br>
$z_1z_2=ac+adj+bcj+bd(−1)$ <br>
$z_1z_2=(ac−bd)+(ad+bc)j$

$z_1j=(a+bj)j$ <br>
$z_1j=aj+bj^2$ <br>
$z_1j=aj+b(−1)$ <br>
$z_1j=(−b)+aj$

$(z_1j)j=(−b+aj)j=−bj+aj^2=−a−bj$ 

# Complex Conjugate 
$z=a+bj$ <br>
$\overline{z}=a−bj$ <br>
$\operatorname{Im}(\overline{z}) = -\operatorname{Im}(z)$

$z\overline{z} = (a+bj)(a-bj)$ <br>
$=a^2 + abj - abj - b^2j^2$ <br>
$=a^2 + (abj-abj) - b^2(-1)$ <br>
$=a^2 + b^2$

# Division
$\frac{z_1}{z_2} = \frac{a+bj}{c+dj}$ <br>
$\frac{z_1}{z_2} = \frac{z_1(\overline{z_2})}{z_2(\overline{z_2})}$ <br>
$\frac{z_1}{z_2} = \frac{(a+bj)(c-dj)}{(c+dj)(c-dj)}$ <br>
$\frac{z_1}{z_2} = \frac{(a+bj)(c-dj)}{c^2 + d^2}$ <br>
$\frac{z_1}{z_2} = \frac{(ac + bd) + (bc-ad)j}{c^2 + d^2}$ <br>
$\frac{z_1}{z_2} = \frac{ac+bd}{c^2 + d^2} + \frac{ac-bd}{c^2 + d^2}j$

# Resolução de Problemas
Se for expressar os números $\frac{1}{2}e^{j\pi}, \frac{1}{2}e^{-j\pi}, e^{\frac{j\pi}{2}}, e^{-\frac{j\pi}{2}}, e^{\frac{j5\pi}{2}}, \sqrt{2}e^{\frac{j\pi}{4}}, \sqrt{2}e^{\frac{j9\pi}{4}}, \sqrt{2}e^{-\frac{j9\pi}{4}}, \sqrt{2}e^{\frac{-9\pi}{4}}$ <br>em forma cartesiana (x+jy) <br>

Para $\frac{1}{2}e^{j\pi}$:<br>
Usando $e^{j\theta} = \cos(\theta) + j\sin(\theta)$ <br>
Torna $e^{j\pi} = \cos(\pi) + j\sin(\pi) \implies e^{j\pi} = 1 + j0$ <br>
E $\frac{1}{2}e^{j\pi} = (-1)\frac{1}{2}$ <br>
Com coordenadas $(-\frac{1}{2},0)$

Para $\frac{1}{2}e^{-j\pi}$:<br>
Usando $e^{j\theta} = \cos(\theta) + j\sin(\theta)$ <br>
Torna $e^{-j\pi} = \cos(-\pi) + j\sin(-\pi) \implies e^{-j\pi} = 1 + j0$ <br>
E $\frac{1}{2}e^{-j\pi} = (-1)\frac{1}{2}$ <br>
Com coordenadas $(-\frac{1}{2},0)$

isso implica que $\frac{1}{2}e^{j\pi} == \frac{1}{2}e^{-j\pi}$

Para $e^{\frac{j\pi}{2}}$:<br>
Usando $e^{j\theta} = \cos(\theta) + j\sin(\theta)$ <br>
Torna $e^{\frac{j\pi}{2}} = \cos(\frac{\pi}{2}) + j\sin(\frac{\pi}{2}) \implies e^{\frac{j\pi}{2}} = j$ <br>
Com coordenadas $(0,1)$

Então: $e^{\frac{j\pi}{2}} = (0,1) \implies e^{\frac{-j\pi}{2}} = (0,1)$

Para $e^{\frac{j5\pi}{2}}$:<br>
Usando a técnica $\phi = \phi_0 + 2k\pi$<br>
Tem se que: $\frac{5\pi}{2} = \frac{\pi}{2} + \frac{4\pi}{2} = \frac{\pi}{2} + 2\pi$ <br>
Usando $e^{j\theta} = \cos(\theta) + j\sin(\theta)$ <br>
E $\cos(a+b) = \cos(a)\cos(b) - \sin(a)\sin(b)$ <br>
E $\sin(a+b) = \sin(a)\cos(b) + \sin(b)\cos(a)$<br>
Então $e^{\frac{\pi}{2} + 2\pi} = \cos(\frac{\pi}{2})\cos(2\pi) - \sin(\frac{\pi}{2})\sin(2\pi) + j(\sin(\frac{\pi}{2})\cos(2\pi) + \sin(2\pi)\cos(\frac{\pi}{2})) = j$<br>
Com coordenadas $(0,1)$
