## Trigonometria Avançada

### Arco Duplo e Redução de Potência
$\sin(2\theta) = 2\sin\theta\cos\theta$ <br>
$\cos(2\theta) = \cos^2\theta - \sin^2\theta = 2\cos^2\theta - 1 = 1 - 2\sin^2\theta$ <br>
$\tan(2\theta) = \frac{2\tan\theta}{1 - \tan^2\theta}$ <br>
$\sin^2\theta = \frac{1 - \cos(2\theta)}{2}$ <br>
$\cos^2\theta = \frac{1 + \cos(2\theta)}{2}$

### Arco Metade ($\theta/2$)
$\sin\left(\frac{\theta}{2}\right) = \pm\sqrt{\frac{1 - \cos\theta}{2}}$ <br>
$\cos\left(\frac{\theta}{2}\right) = \pm\sqrt{\frac{1 + \cos\theta}{2}}$ <br>
$\tan\left(\frac{\theta}{2}\right) = \frac{1 - \cos\theta}{\sin\theta} = \frac{\sin\theta}{1 + \cos\theta}$

### Transformações de Prostaférese (Produto <-> Soma)
$\sin a \cos b = \frac{1}{2}[\sin(a+b) + \sin(a-b)]$ <br>
$\cos a \cos b = \frac{1}{2}[\cos(a+b) + \cos(a-b)]$ <br>
$\sin a \sin b = \frac{1}{2}[\cos(a-b) - \cos(a+b)]$ <br>

$\sin x + \sin y = 2\sin\left(\frac{x+y}{2}\right)\cos\left(\frac{x-y}{2}\right)$ <br>
$\sin x - \sin y = 2\sin\left(\frac{x-y}{2}\right)\cos\left(\frac{x+y}{2}\right)$ <br>
$\cos x + \cos y = 2\cos\left(\frac{x+y}{2}\right)\cos\left(\frac{x-y}{2}\right)$ <br>
$\cos x - \cos y = -2\sin\left(\frac{x+y}{2}\right)\sin\left(\frac{x-y}{2}\right)$

### Substituição de Weierstrass ($t = \tan(\theta/2)$)
$\sin\theta = \frac{2t}{1 + t^2} \quad \cos\theta = \frac{1 - t^2}{1 + t^2} \quad \tan\theta = \frac{2t}{1 - t^2}$

### Formas Complexas (Euler & de Moivre)
$e^{i\theta} = \cos\theta + i\sin\theta$ <br>
$\cos\theta = \frac{e^{i\theta} + e^{-i\theta}}{2} \quad \sin\theta = \frac{e^{i\theta} - e^{-i\theta}}{2i}$ <br>
$(\cos\theta + i\sin\theta)^n = \cos(n\theta) + i\sin(n\theta)$

## Limites Trigonométricos Fundamentais

* **Limite Fundamental do Seno:** <br>
$\lim_{x \to 0} \frac{\sin(x)}{x} = 1 \implies \lim_{u(x) \to 0} \frac{\sin(u)}{u} = 1$

* **Limite Fundamental da Tangente:**<br>
$\lim_{x \to 0} \frac{\tan(x)}{x} = 1$

* **Limite do Cosseno:**<br>
$\lim_{x \to 0} \frac{1 - \cos(x)}{x} = 0 \quad \text{e} \quad \lim_{x \to 0} \frac{1 - \cos(x)}{x^2} = \frac{1}{2}$

## Derivadas Trigonométricas

### Formato Simples 
$\frac{d}{dx}\sin(x) = \cos(x)$<br>
$\frac{d}{dx}\cos(x) = -\sin(x)$<br>
$\frac{d}{dx}\tan(x) = \sec^2(x)$<br>
$\frac{d}{dx}\cot(x) = -\csc^2(x)$<br>
$\frac{d}{dx}\sec(x) = \sec(x)\tan(x)$<br>
$\frac{d}{dx}\csc(x) = -\csc(x)\cot(x)$<br>

### Formato Composto (Regra da Cadeia: $u = g(x)$)
$\frac{d}{dx}\sin(u) = \cos(u) \cdot u'$<br>
$\frac{d}{dx}\cos(u) = -\sin(u) \cdot u'$<br>
$\frac{d}{dx}\tan(u) = \sec^2(u) \cdot u'$<br>
$\frac{d}{dx}\cot(u) = -\csc^2(u) \cdot u'$<br>
$\frac{d}{dx}\sec(u) = \sec(u)\tan(u) \cdot u'$<br>
$\frac{d}{dx}\csc(u) = -\csc(u)\cot(u) \cdot u'$<br>

## Derivadas Trigonométricas Inversas

$\frac{d}{dx}\arcsin(u) = \frac{u'}{\sqrt{1-u^2}}$<br>
$\frac{d}{dx}\arccos(u) = \frac{-u'}{\sqrt{1-u^2}}$<br>
$\frac{d}{dx}\arctan(u) = \frac{u'}{1+u^2}$<br>
$\frac{d}{dx}\operatorname{arccot}(u) = \frac{-u'}{1+u^2}$<br>
$\frac{d}{dx}\operatorname{arcsec}(u) = \frac{u'}{|u|\sqrt{u^2-1}}$<br>
$\frac{d}{dx}\operatorname{arccsc}(u) = \frac{-u'}{|u|\sqrt{u^2-1}}$<br>

## Integrais Trigonométricas

### Integrais Diretas
$\int \sin(x) \, dx = -\cos(x) + C$ <br>
$\int \cos(x) \, dx = \sin(x) + C$ <br>
$\int \tan(x) \, dx = \ln|\sec(x)| + C$ <br>
$\int \cot(x) \, dx = \ln|\sin(x)| + C$ <br>
$\int \sec(x) \, dx = \ln|\sec(x) + \tan(x)| + C$ <br>
$\int \csc(x) \, dx = -\ln|\csc(x) + \cot(x)| + C$ <br>

### Integrais de Potências e Produtos Notáveis
$\int \sec^2(x) \, dx = \tan(x) + C$ <br>
$\int \csc^2(x) \, dx = -\cot(x) + C$ <br>
$\int \sec(x)\tan(x) \, dx = \sec(x) + C$ <br>
$\int \csc(x)\cot(x) \, dx = -\csc(x) + C$ <br>


## Integrais que Resultam em Funções Inversas

$\int \frac{1}{\sqrt{a^2-x^2}} \, dx = \arcsin\left(\frac{x}{a}\right) + C$ <br>
$\int \frac{1}{a^2+x^2} \, dx = \frac{1}{a}\arctan\left(\frac{x}{a}\right) + C$ <br>
$\int \frac{1}{x\sqrt{x^2-a^2}} \, dx = \frac{1}{a}\operatorname{arcsec}\left(\frac{|x|}{a}\right) + C$ <br>

# Funções Hiperbólicas e Identidades Avançadas no Cálculo

## Definições Algébricas e Identidades Fundamentais

### Definições Exponenciais
$\sinh(x) = \frac{e^x - e^{-x}}{2} \quad \cosh(x) = \frac{e^x + e^{-x}}{2}$ <br>
$\tanh(x) = \frac{\sinh(x)}{\cosh(x)} = \frac{e^x - e^{-x}}{e^x + e^{-x}} \quad \coth(x) = \frac{\cosh(x)}{\sinh(x)} = \frac{e^x + e^{-x}}{e^x - e^{-x}}$<br>
$\operatorname{sech}(x) = \frac{1}{\cosh(x)} = \frac{2}{e^x + e^{-x}} \quad \operatorname{csch}(x) = \frac{1}{\sinh(x)} = \frac{2}{e^x - e^{-x}}$<br>

### Identidades Hiperbólicas Fundamentais
$\cosh^2(x) - \sinh^2(x) = 1$ <br>
$1 - \tanh^2(x) = \operatorname{sech}^2(x)$<br>
$\coth^2(x) - 1 = \operatorname{csch}^2(x)$<br>

### Soma, Subtração e Arco Duplo
$\sinh(a \pm b) = \sinh(a)\cosh(b) \pm \cosh(a)\sinh(b)$<br>
$\cosh(a \pm b) = \cosh(a)\cosh(b) \pm \sinh(a)\sinh(b)$<br>
$\sinh(2x) = 2\sinh(x)\cosh(x)$<br>
$\cosh(2x) = \cosh^2(x) + \sinh^2(x) = 2\cosh^2(x) - 1 = 1 + 2\sinh^2(x)$<br>


## Derivadas Hiperbólicas

### Formato Simples
$\frac{d}{dx}\sinh(x) = \cosh(x)$<br>
$\frac{d}{dx}\cosh(x) = \sinh(x)$<br>
$\frac{d}{dx}\tanh(x) = \operatorname{sech}^2(x)$<br>
$\frac{d}{dx}\coth(x) = -\operatorname{csch}^2(x)$<br>
$\frac{d}{dx}\operatorname{sech}(x) = -\operatorname{sech}(x)\tanh(x)$<br>
$\frac{d}{dx}\operatorname{csch}(x) = -\operatorname{csch}(x)\coth(x)$<br>

### Formato Composto (Regra da Cadeia: $u = g(x)$)
$\frac{d}{dx}\sinh(u) = \cosh(u) \cdot u'$<br>
$\frac{d}{dx}\cosh(u) = \sinh(u) \cdot u'$<br>
$\frac{d}{dx}\tanh(u) = \operatorname{sech}^2(u) \cdot u'$<br>
$\frac{d}{dx}\coth(u) = -\operatorname{csch}^2(u) \cdot u'$<br>
$\frac{d}{dx}\operatorname{sech}(u) = -\operatorname{sech}(u)\tanh(u) \cdot u'$<br>
$\frac{d}{dx}\operatorname{csch}(u) = -\operatorname{csch}(u)\coth(u) \cdot u'$<br>


## Derivadas Hiperbólicas Inversas

$\frac{d}{dx}\operatorname{arcsinh}(u) = \frac{u'}{\sqrt{u^2 + 1}}$<br>
$\frac{d}{dx}\operatorname{arccosh}(u) = \frac{u'}{\sqrt{u^2 - 1}} \quad (u > 1)$<br>
$\frac{d}{dx}\operatorname{arctanh}(u) = \frac{u'}{1 - u^2} \quad (|u| < 1)$<br>
$\frac{d}{dx}\operatorname{arccoth}(u) = \frac{u'}{1 - u^2} \quad (|u| > 1)$<br>
$\frac{d}{dx}\operatorname{arcsech}(u) = \frac{-u'}{u\sqrt{1 - u^2}} \quad (0 < u < 1)$<br>
$\frac{d}{dx}\operatorname{arccsch}(u) = \frac{-u'}{|u|\sqrt{1 + u^2}} \quad (u \neq 0)$<br>

## Integrais Hiperbólicas

### Integrais Diretas
$\int \sinh(x) \, dx = \cosh(x) + C$<br>
$\int \cosh(x) \, dx = \sinh(x) + C$<br>
$\int \tanh(x) \, dx = \ln(\cosh(x)) + C$<br>
$\int \coth(x) \, dx = \ln|\sinh(x)| + C$<br>
$\int \operatorname{sech}(x) \, dx = \arctan(\sinh(x)) + C$<br>
$\int \operatorname{csch}(x) \, dx = \ln\left|\tanh\left(\frac{x}{2}\right)\right| + C$<br>

### Integrais de Potências e Produtos Notáveis
$\int \operatorname{sech}^2(x) \, dx = \tanh(x) + C$<br>
$\int \operatorname{csch}^2(x) \, dx = -\coth(x) + C$<br>
$\int \operatorname{sech}(x)\tanh(x) \, dx = -\operatorname{sech}(x) + C$<br>
$\int \operatorname{csch}(x)\coth(x) \, dx = -\operatorname{csch}(x) + C$<br>

## Integrais Racionais e Substituição Hiperbólica

Formulário fundamental usado em substituição hiperbólica para eliminar radicais do tipo $\sqrt{x^2+a^2}$ e $\sqrt{x^2-a^2}$:

$\int \frac{1}{\sqrt{x^2 + a^2}} \, dx = \operatorname{arcsinh}\left(\frac{x}{a}\right) + C = \ln\left|x + \sqrt{x^2 + a^2}\right| + C$<br>
$\int \frac{1}{\sqrt{x^2 - a^2}} \, dx = \operatorname{arccosh}\left(\frac{x}{a}\right) + C = \ln\left|x + \sqrt{x^2 - a^2}\right| + C$<br>
$\int \frac{1}{a^2 - x^2} \, dx = \frac{1}{a}\operatorname{arctanh}\left(\frac{x}{a}\right) + C = \frac{1}{2a}\ln\left|\frac{a + x}{a - x}\right| + C \quad (|x| < a)$<br>
