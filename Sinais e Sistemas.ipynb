# Sinais
O sinal pode ser conhecido com 3 modos.
1. Sinal é um fenômeno físico que varia com o tempo e tem como objetivo transmitir informações. 
2. O sinal é uma função do tempo.
3. Um sinal é uma função de uma ou mais variáveis ​​independentes, que contêm alguma informação.

**Nota**: O ruído também é um sinal, mas a informação transmitida por ele é indesejável, sendo, portanto, considerada inadequada. 



# Sistema
Um sistema é um dispositivo ou combinação de dispositivos que pode operar com base em sinais e produzir uma resposta correspondente. <br>
A entrada em um sistema é chamada de excitação e a saída dele é chamada de resposta. 

Para uma ou mais entradas, o sistema pode ter uma ou mais saídas. 



# Função degrau unitário 
Representa uma mudança abrupta de estado, funcionando como um interruptor ideal que é ligado no instante $t = 0$.

$u(t) = \begin{cases} 1 & t \geqslant 0 \\ 0 & t < 0 \end{cases}$



Aplicação: Permite "acender" ou habilitar sinais a partir de um tempo específico, delimitando o início de uma observação <br>
matemática ou simulação computacional.

# Impulso Unitário ou Delta de Dirac ($\delta(t)$)
É uma idealização matemática caracterizada por uma duração infinitamente curta, altura infinita,<br>
mas com área total sob a curva exatamente igual a $1$.

$\delta(t) = 0 \text{ para } t \neq 0, \quad \int_{-\infty}^{\infty} \delta(t) \, dt = 1$



- Relação: É formalmente a derivada matemática da função degrau unitário: $\frac{du(t)}{dt} = \delta(t)$.
- Aplicação: Utilizado para encontrar a resposta impulsiva de sistemas lineares e invariantes no tempo (LIT),
representando um choque mecânico ou um pulso elétrico instantâneo.

# Sinal de Rampa ($r(t)$)
Representa um crescimento linear e constante ao longo do tempo.

$r(t) = \begin{cases} t & t \ge 0 \\ 0 & t < 0 \end{cases}$



- Relação: É o resultado da integral da função degrau no tempo: $r(t) = \int_{-\infty}^{t} u(\tau) \, d\tau$.
- Aplicação: Empregado para modelar sistemas que sofrem aceleração constante, rampas de tensão em geradores de <br>
varredura ou erros acumulativos em sistemas de controle.


# Sinal Parabólico (p(t))
- Fórmula: $p(t) = \frac{t^2}{2} u(t)$ (ou $t^2 u(t)$ dependendo da convenção).
- Função: Cresce de forma quadrática (curva parabólica) a partir de $t = 0$. É o resultado da integração contínua do sinal de rampa, útil para modelar sistemas com aceleração constante ou erros parabólicos de rastreamento.



# Função Sinal ($\text{sgn}(t)$)
- Fórmula: $\text{sgn}(t) = \begin{cases} 1 & t > 0 \\ 0 & t = 0 \\ -1 & t < 0 \end{cases}$
- Função: Retorna o sinal algébrico de uma variável. É amplamente utilizada em modulações digitais (como chaveamento de fase) e na teoria de controle para representar saturações simétricas.


# Sinal Exponencial ($x(t)$)
- Fórmula: $x(t) = e^{at} u(t)$
- Função: Modela crescimentos ($a > 0$) ou decaimentos ($a < 0$) exponenciais. Essencial para descrever a carga e descarga de capacitores, respostas transitórias de circuitos RLC e o fator de amortecimento de sistemas físicos.

<br>
<br>


# Sinal Retangular ($rect(t)$ ou $\Pi(t)$)
- Fórmula: $\text{rect}(t) = \begin{cases} 1 & \vert{}t\vert{} \le 1/2 \\ 0 & \vert{}t\vert{} > 1/2 \end{cases}$
- Função: Representa um pulso de largura finita e amplitude unitária centrado na origem. Na comunicação, modela bits digitais em banda base e janelas de observação temporal.


# Sinal Triangular ($tri(t)$ ou $\Lambda(t)$)
- Fórmula: $\text{tri}(t) = \begin{cases} 1 - \vert{}t\vert{} & \vert{}t\vert{} \le 1 \\ 0 & \vert{}t\vert{} > 1 \end{cases}$
- Função: Uma forma de onda em pico linear que decresce simetricamente. É muito usada em processamento de sinais como função de janela ou para gerar rampas simétricas.



# Sinal Sinusoidal
- Fórmula: $x(t) = A \cos(2\pi f t + \phi)$
- Função: A base de toda a análise espectral e telecomunicações. Representa uma oscilação harmônica pura que, pelo Teorema de Fourier, serve como bloco de construção para qualquer sinal periódico complexo.



# Função Sinc ($\text{sinc}(t)$)
- Fórmula: $\text{sinc}(t) = \frac{\sin(\pi t)}{\pi t}$
- Função: Uma onda oscilatória cuja amplitude decresce simetricamente à medida que se afasta da origem. É a transformada de Fourier exata de um pulso retangular e dita a largura de banda teórica de canais de transmissão.



# Função de Amostragem ($\text{sa}(t)$)
- Fórmula: $\text{sa}(t) = \frac{\sin(t)}{t}$
- Função: Variação da função sinc (com o fator $\pi$ omitido). Aparece de forma natural em problemas de amostragem de sinais e teoria de antenas (padrões de radiação).



# Classificação de Sinais
**Categoria**:
## Sinais de tempo contínuo e sinais de tempo discreto 
Diz-se que um sinal é contínuo quando está definido para todos os instantes de tempo. <br>
<br>
Diz-se que um sinal é discreto quando é definido apenas em instantes de tempo discretos. <br>
<br>

# Sinais determinísticos e não determinísticos 
Um sinal é considerado determinístico se não houver incerteza quanto ao seu valor em qualquer <br>
instante de tempo. Ou seja, sinais que podem ser definidos exatamente por uma fórmula matemática <br>
são conhecidos como sinais determinísticos. 



Um sinal é considerado não determinístico se houver incerteza quanto ao seu valor em um determinado <br>
instante. Sinais não determinísticos são aleatórios por natureza, daí o nome de sinais aleatórios. <br>
Sinais aleatórios não podem ser descritos por uma equação matemática. Eles são modelados em termos <br>
probabilísticos. 



# Sinais Pares e Ímpares
**Função Par (Sinal Par)**
- **Definição Matemática**: $x(-t) = x(t)$
- **Comportamento Gráfico**: O sinal é perfeitamente espelhado em relação ao eixo vertical (eixo da ordenada). <br>
Se você inverter o tempo ($t$ para $-t$), a amplitude permanece exatamente igual.
- **Exemplos Práticos**: A função cosseno $\cos(t)$, o sinal parabólico $t^2$ e o pulso retangular centralizado <br>
$\text{rect}(t)$.

**Função Ímpar (Sinal Ímpar)**
- **Definição Matemática**: $x(-t) = -x(t)$
- **Comportamento Gráfico**: O sinal possui simetria rotacional (ou central) em relação à origem $(0,0)$. Inverter <br>
o tempo inverte o sinal da amplitude.
- **Exemplos Práticos**: A função seno $\sin(t)$, a função sinal $\text{sgn}(t)$ e o sinal de rampa $t$.

## Decomposição de Sinais
Qualquer sinal arbitrário e aperiódico $x(t)$ que não seja estritamente par ou ímpar pode ser matematicamente <br>
separado na soma de uma componente par $x_e(t)$ e uma componente ímpar $x_o(t)$:
$x(t) = x_e(t) + x_o(t)$<br>

As fórmulas para extrair essas componentes são:

$x_e(t) = \frac{1}{2}[x(t) + x(-t)]$<br>
$x_0(t) = \frac{1}{2}[x(t) - x(-t)]$

# Sinais Periodicos e Aperiodicos 
A classificação fundamental que dita como um sinal se comporta ao longo do tempo e como ele se traduz no <br>
domínio da frequência.

**Sinais Periódicos**:
- Definição Matemática: Um sinal $x(t)$ é periódico se existe um valor constante e positivo $T_0$ tal que: <br>
$x(t) = x(t + T_0) \quad \text{para todo } t$
- Período Fundamental ($T_0$): É o menor valor positivo de tempo para o qual a propriedade de repetição <br>
se cumpre. A frequência fundamental associada é $f_0 = \frac{1}{T_0}$.
- Comportamento: Repete exatamente o mesmo padrão de ciclos infinitamente ao longo do tempo.
- Exemplos: Funções senoidais puras, ondas quadradas, ondas triangulares e exponenciais complexas da forma <br>
$e^{j \omega_0 t}$.
- Espectro de Frequência: Geram espectros discretos compostos por linhas isoladas (frequências fundamentais e suas harmônicas) quando analisados via Série de Fourier.

**Sinais Aperiódicos**:
- Definição Matemática: São sinais que não possuem um padrão que se repete em intervalos regulares de tempo, ou seja, não existe um $T_0$ finito que satisfaça a condição de periodicidade.
- Comportamento: Podem ocorrer de forma transitória (como um pulso único que surge e desaparece) ou apresentar variações estocásticas/aleatórias contínuas.
- Exemplos: O degrau unitário $u(t)$, o impulso $\delta(t)$, o pulso retangular isolado $\text{rect}(t)$, sinais de voz humana e ruídos térmicos.
- Espectro de Frequência: Produzem espectros contínuos (uma faixa contínua e densa de frequências) quando analisados via Transformada de Fourier.

**A Transição Analítica**
Na prática de engenharia, um sinal aperiódico de duração finita pode ser matematicamente encarado como o caso limite de um sinal periódico cujo período T0​ tende ao infinito. Essa transição conceitual é exatamente a ponte que transforma a Série de Fourier (para sinais periódicos) na Transformada de Fourier (para sinais aperiódicos).

# Sinais de Energia e Potência
**Sinais de Energia**
- Definição: São sinais que possuem energia total finita e não nula ($0 < E < \infty$). Para que isso aconteça, o sinal precisa ter uma duração temporal limitada (sinais transitórios).
- Potência Média: Como a energia está concentrada em uma janela finita, quando diluída ao longo de um tempo infinito, a potência média resulta em zero ($P = 0$).
- Fórmula da Energia: $E = \int_{-\infty}^{\infty} \vert{}x(t)\vert{}^2 \, dt$
- Exemplos Práticos: O pulso retangular isolado $\text{rect}(t)$, uma exponencial decrescente $e^{-at}u(t)$ ou qualquer transiente de laboratório que começa e termina.

**Sinais de Potência**
- Definição: São sinais que possuem energia total infinita ($E = \infty$), mas cuja potência média por unidade de tempo é finita e diferente de zero ($0 < P < \infty$).
- Comportamento: Persistem indefinidamente no tempo (geralmente sinais estocásticos ou periódicos que nunca cessam).
- Fórmula da Potência Média: $P = \lim_{T \to \infty} \frac{1}{T} \int_{-T/2}^{T/2} \vert{}x(t)\vert{}^2 \, dt$
- Exemplos Práticos: Funções senoidais puras, ondas quadradas periódicas, o degrau unitário $u(t)$ e sinais de portadora em sistemas de comunicação.

**Regra de Exclusão**
Um sinal nunca pode pertencer a ambas as categorias simultaneamente: ou ele tem energia finita (e potência zero), ou tem potência finita (e energia infinita). Além disso, existem sinais que não pertencem a nenhuma das duas classes (como o sinal de rampa $r(t)$), pois tanto a sua energia quanto a sua potência divergem para o infinito.

# Sinais Reais e Imaginários
A classificação entre sinais reais e complexos (que englobam componentes imaginárias) define o domínio numérico em que a amplitude do sinal opera a cada instante de tempo.

**Sinais Reais**
- Definição: A amplitude do sinal em qualquer instante $t$ é um número estritamente real, ou seja, $x(t) \in \mathbb{R}$.
- Comportamento: São os sinais que podem ser diretamente medidos por instrumentos físicos no laboratório, como um osciloscópio ou um multímetro.
- Exemplos: Uma tensão senoidal pura na tomada ($\cos(\omega t)$), a voz capturada por um microfone ou um pulso retangular de voltagem.
- Propriedade Espectral: O espectro de frequência de um sinal real possui simetria hermitiana, o que significa que as frequências negativas são o conjugado complexo das frequências positivas.

**Sinais Complexos**
- Definição: São sinais cuja amplitude em cada instante $t$ é um número complexo, representados algebricamente na forma: $x(t) = x_r(t) + j x_i(t)$ <br>
onde $x_r(t)$ é a parte real, $x_i(t)$ é a parte imaginária e $j = \sqrt{-1}$ (unidade imaginária).
- A Ferramenta Suprema: O exemplo mais importante em engenharia é a exponencial complexa fundamentada na identidade de Euler: $e^{j\omega t} = \cos(\omega t) + j\sin(\omega t)$
- Utilidade Prática: Embora não existam fios conduzindo "correntes imaginárias" no mundo físico, os sinais complexos são indispensáveis na modelagem matemática. Eles simplificam drasticamente o cálculo de circuitos em regime permanente AC (fasores), análises de modulação em quadratura (QAM) e o manuseio algébrico da Transformada de Fourier.

# Operações de Sinais
1. **Operações na Amplitude (Eixo Vertical)** <br>
Essas operações alteram o "tamanho" ou combinam os valores de voltagem/corrente do sinal em cada instante, sem mexer na linha do tempo.
- Escala de Amplitude (Amplitude Scaling): <br>
  Multiplicar o sinal por uma constante $C$.<br>
  Exemplo: $y(t) = C \cdot x(t)$. Se $C = 2$, a onda dobra de altura (amplificação). Se $C = 0.5$, ela é atenada.

- Adição e Subtração de Sinais:<br>
  Somar ou subtrair as amplitudes de dois sinais ponto a ponto no mesmo instante de tempo. <br>
  Exemplo: $z(t) = x_1(t) + x_2(t)$. Se no instante $t=1$ o sinal 1 vale 3 e o sinal 2 vale 2, o resultado terá amplitude $3 + 2 = 5$.

- Multiplicação de Sinais: <br>
  Multiplicar as amplitudes de dois sinais em cada instante.<br>
  Aplicação crucial: É exatamente isso que acontece na Modulação AM! Você multiplica a onda de mensagem de baixa <br>
  frequência pela portadora de alta frequência para conseguir transmiti-la pelo ar.


2. **Operações no Tempo (Eixo Horizontal)**<br>
Essas operações modificam a forma como o sinal se distribui ao longo da linha do tempo. Elas exigem um pouco de atenção porque o comportamento algébrico costuma ser o "oposto" do que nossa intuição imagina visualmente.
- Deslocamento Temporal (Time Shifting): <br>
  Atrasa ou adianta o sinal no tempo. <br>
  $x(t - t_0)$ com $t_0 > 0$: Move o sinal para a direita (atraso). Se era para acontecer em $t=0$, agora acontece em $t = t_0$.<br>
  $x(t + t_0)$ com $t_0 > 0$: Move o sinal para a esquerda (avanço/antecipação).
- Escala Temporal (Time Scaling):<br>
  Acelera ou desacelera o sinal.
  $x(A \cdot t)$: <br>
      Se $\vert{}A\vert{} > 1$, o sinal sofre compressão temporal (é reproduzido mais rápido, como um áudio acelerado). <br>
      Se $\vert{}A\vert{} < 1$, o sinal sofre expansão temporal (é esticado, como um câmera lenta).

- Inversão Temporal (Time Reversal): <br>
  Ocorre quando fazemos $x(-t)$. O sinal é completamente espelhado horizontalmente (o que estava no futuro vai para o passado, como rodar uma fita de vídeo ao contrário).  

# Sistemas

1. **Sistemas Lineares e Não-Lineares (Linear and Non-Linear)**
   - Linear: Precisa obedecer simultaneamente aos princípios de homogeneidade (se a entrada for multiplicada por uma constante, a saída é multiplicada pela mesma constante) e superposição (a resposta a uma soma de entradas é a soma das respostas individuais).
   - Não-Linear: Apresenta efeitos como saturação, distorções ou multiplicações entre sinais (como os moduladores AM que vimos anteriormente).
2. **Sistemas Invariantes e Variantes no Tempo (Time-Invariant and Time-Variant)**
   - Invariante no Tempo (TI): As características físicas do sistema não mudam com o passar do tempo. Se você aplicar um sinal hoje ou daqui a uma semana, a resposta do sistema será exatamente a mesma, exceto por um atraso correspondente.
   - Variante no Tempo (TV): Os parâmetros do sistema mudam com o tempo (por exemplo, um circuito cujos componentes envelhecem ou mudam de valor ativamente). <br>
     Destaque: A união dos dois mais importantes dá origem aos sistemas LTI (Linear Time-Invariant), que formam a espinha dorsal de quase toda a teoria de circuitos, transformadas de Fourier e engenharia de comunicações.
3. **Sistemas Estáticos e Dinâmicos (Static and Dynamic)**
   - Estáticos (Sem Memória): A saída em qualquer instante depende apenas da entrada aplicada naquele exato instante. Exemplo: um resistor simples regido pela Lei de Ohm ($v(t) = R \cdot i(t)$).
   - Dinâmicos (Com Memória): A saída depende de entradas passadas ou futuras. Exemplo: capacitores e indutores, que armazenam energia e dependem de integrais ou derivadas do sinal ao longo do tempo.
4. **Sistemas Causais e Não-Causais (Causal and Non-Causal)**
   - Causais: A saída depende apenas de valores presentes e passados da entrada. Nenhum sistema físico real construído no nosso universo pode olhar para o futuro.
   - Não-Causais: A saída depende de valores futuros da entrada. Eles só existem de forma teórica ou em processamento de sinais gravados (onde o "futuro" do sinal já está armazenado em um arquivo de computador e pode ser lido).
5. **Sistemas Estáveis e Instáveis (Stable and Unstable)**
   - Estáveis (Critério BIBO - Bounded-Input, Bounded-Output): Se você aplicar uma entrada limitada (com amplitude finita), a saída gerada também terá amplitude finita.
   - Instáveis: Uma entrada perfeitamente controlada faz a saída crescer indefinidamente rumo ao infinito (como um sistema com realimentação positiva descontrolada ou um oscilador ideal sem amortecimento).
