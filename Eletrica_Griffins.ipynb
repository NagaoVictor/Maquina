# Características de um Vetor

## 1. Módulo (Magnitude ou Tamanho)

$$\vec{A} = A_x\hat{i} + A_y\hat{j} + A_z\hat{k}$$
- O módulo representa o comprimento geométrico do vetor. Ele é sempre um número real não negativo.
- Na Álgebra Cartesiana: É calculado generalizando o Teorema de Pitágoras para três dimensões:
$$\vert{}\vec{A}\vert{} = A = \sqrt{A_x^2 + A_y^2 + A_z^2}$$
Significado Físico: Na eletrodinâmica, o módulo de um vetor de campo (como o campo elétrico $\vert{}\vec{E}\vert{}$) representa a "intensidade" da força que uma carga experimentará naquele ponto.

2. Direção e Sentido
- Direção: A linha reta ao longo da qual o vetor se estende (ou um conjunto de retas paralelas a ela).
- Sentido: Para qual dos dois lados da linha a seta aponta (indicado pela ponta da seta).
- A Ferramenta Matemática (Vetor Unitário): Para separar a "direção e sentido" do "tamanho", o Griffiths utiliza o vetor unitário ($\hat{a}$), cujo módulo é exatamente $1$:$$\hat{a} = \frac{\vec{A}}{\vert{}\vec{A}\vert{}}$$Isso permite reescrever qualquer vetor como o produto de sua magnitude pelo seu vetor unitário: $\vec{A} =\vert{}\vec{A}\vert{}\hat{a}$.

## 1.1 Álgebra Vetorial: Produto Escalar e Vetorial

Na eletrodinâmica, os campos dependem fortemente da geometria espacial. Precisamos dominar as duas principais operações entre vetores $\vec{A}$ e $\vec{B}$:

1. **Produto Escalar (Dot Product):** Resulta em um escalar. Mede a projeção de um vetor sobre o outro. <br>
   $\vec{A} \cdot \vec{B} = A_x B_x + A_y B_y + A_z B_z = AB \cos\theta$ <br>
   $a.b = ||a||.||b||.\cos{\theta}$ <br>
   $a.b = 1$ = Ortonormal, significa que os vetores estão paralelos <br>
   $a.b = 0$ = Ortogonal, significa que os vetores são perpendiculares <br>
   
   $\hat{x} \cdot \hat{x} = \hat{y} \cdot \hat{y} = \hat{z} \cdot \hat{z} = 1$<br>
   $\hat{x} \cdot \hat{y} = \hat{y} \cdot \hat{z} = \hat{z} \cdot \hat{x} = 0$<br>
2. **Produto Vetorial (Cross Product):** Resulta em um novo vetor ortogonal (perpendicular) ao plano formado por $\vec{A}$ e $\vec{B}$. <br>
   $\vec{A} \times \vec{B} = \det \begin{vmatrix} \hat{i} & \hat{j} & \hat{k} \\ A_x & A_y & A_z \\ B_x & B_y & B_z \end{vmatrix}$ <br>
   $a.b = a.b.\sin{\theta}.\hat{n}$ <br>
   
3. **Produto Tríplo:** Resulta em um volume de um paralelepípedo <br>
   $\vec{A} \times \vec{B} = \det \begin{vmatrix} u_x & u_y & u_z \\ v_x & v_y & v_z \\ w_x & w_y & w_z \end{vmatrix}$ <br>
   $u \times (v \times w) = (u \cdot w)\cdot v -  (u \cdot v)\cdot w$ <br>
   

# O Operador $\nabla$

$\nabla = \hat{x}\frac{𝜕}{𝜕x} + \hat{y}\frac{𝜕}{𝜕y} + \hat{z}\frac{𝜕}{𝜕z}$ <br>

*Operações básicas*:
- Multiplicar um escalar $\alpha: A\alpha$
- Multiplicar outro vetor B, através do produto escalar: $A \cdot B$
- Multiplicar outro vetor através do produto vetorial: $A \times B$

*Analogamente*:
- Em uma função $T: \nabla T$ (Gradiente)
- Em uma função vetorial v, através do produto escalar: $\nabla \cdot v$ (Divergente)
- Em uma função vetorial v, através do produto vetorial: $\nabla \times v$ (Rotacional)

1. O Gradiente ($\nabla T$): A Translação DirecionalO gradiente atua sobre um campo escalar e devolve um vetor. Ele não tem nada a ver com rotação pura; ele lida estritamente com direção e inclinação translacional.A natureza: É um vetor puramente translacional. Ele diz: "Se você quiser subir o morro o mais rápido possível, dê um passo nesta direção linear exata". Ele aponta para onde a grandeza mais cresce no espaço.

2. O Divergente ($\nabla \cdot \vec{v}$): A Translação Radial (Expansão/Compressão)O divergente mede fontes e sumidouros. Pense no ato de encher um balão de ar ou comprimir um gás. As moléculas estão se movendo para fora (ou para dentro) de forma linear/radial.A natureza: Continua sendo de natureza translacional, mas com foco em mudança de volume. O divergente não quer saber se o fluido está girando em redemoinho; ele quer saber se o fluido está se afastando de um ponto (expansão translacional para fora) ou se acumulando ali (compressão translacional para dentro).

3. O Rotacional ($\nabla \times \vec{v}$): A Natureza Rotacional PuraAqui a chave vira totalmente. O rotacional é o único dos três itens criado especificamente para isolar o giro.A natureza: É a natureza rotacional. Enquanto o divergente ignora o giro e mede apenas a expansão, o rotacional ignora a expansão em linha reta e mede o quanto o campo está forçando um objeto a girar ao redor do próprio eixo (como a roda de pás ou a força magnética fechada ao redor da bobina).

## Gradiente
$\nabla T(x, y, z) = \frac{𝜕T}{𝜕x}\hat{x} + \frac{𝜕T}{𝜕y}\hat{y} + \frac{𝜕T}{𝜕z}\hat{z}$
- Derivada parcial de um vetor T: $\begin{bmatrix} \frac{𝜕T}{𝜕x} & \frac{𝜕T}{𝜕y} & \frac{𝜕T}{𝜕z} \end{bmatrix}$ <br>
Exemplo:<br>
1 - $f(x,y,z) = x^2 + y^3 + z^4$ $\to \nabla f(x,y,z) = \begin{bmatrix} (2x + y^3 + z^4) & (x^2 + 3y^2 + z^4) & (x^2 + y^3 + 4z^3) \end{bmatrix}$ <br>
2 - $f(x,y,z) = x^2y^3z^4$ $\to \nabla f(x,y,z) = \begin{bmatrix} (2xy^2z^3) & (3x^2y^2z^4) & (4x^2y^3z^3) \end{bmatrix}$<br>
3 - $f(x,y,z) = e^x\sin{y}\ln{z}$ $\to \nabla f(x,y,z) = \begin{bmatrix} (e^x\sin{y}\ln{z}) & (e^x\cos{y}\ln{z}) & (\frac{e^x\cos{y}}{z}) \end{bmatrix}$

### A Direção Máxima Subida (Taxa de Variação Máxima)
Se você está em um ponto $(x, y, z)$ em uma superfície escalar e quer saber para qual direção exata deve dar um passo para subir o mais rápido possível, o gradiente aponta exatamente para lá.
- A Matemática por trás: A derivada direcional de $T$ na direção de um vetor unitário $\hat{u}$ é dada por: <br>
$\frac{dT}{ds} = \nabla T \cdot \hat{u} = \vert{}\nabla T\vert{} \cos\theta$ <br>
Onde $\theta$ é o ângulo entre o gradiente e a direção escolhida.
- Como o cosseno atinge seu valor máximo ($1$) quando $\theta = 0^\circ$, conclui-se que o gradiente aponta estritamente na direção onde a função cresce mais rápido. Se você caminhar na direção contrária ($-\nabla T$), você encontra a descida mais íngreme (princípio básico do algoritmo de Gradient Descent em otimização e aprendizado de máquina).

### Ortogonalidade Absoluta às Superfícies de Nível
- Superfície de Nível: É o conjunto de pontos onde a função escalar é constante, ou seja, $T(x, y, z) = C$ (como as linhas de contorno de altitude em um mapa topográfico, onde a altura é a mesma em toda a linha).
- A Propriedade: O vetor gradiente $\nabla T$ em qualquer ponto é sempre perfeitamente ortogonal (perpendicular) à superfície de nível que passa por aquele ponto.
- Utilidade na Elétrica: É daqui que tiramos a relação fundamental entre o Potencial Elétrico ($V$) e o Campo Elétrico ($\vec{E}$). As superfícies onde o potencial é constante chamam-se superfícies equipotenciais. Como o campo elétrico é o gradiente negativo do potencial ($\vec{E} = -\nabla V$), as linhas do campo elétrico cruzam as superfícies equipotenciais sempre a $90^\circ$ (perpendicularmente).

### O Módulo do Gradiente mede a "Inclinação"
Enquanto a direção do gradiente diz para onde ir, o seu módulo ($\vert{}\nabla T\vert{}$) quantifica a intensidade da mudança.
- Ele mede o quão íngreme é a subida naquele ponto específico.
- Se $\vert{}\nabla T\vert{} = 0$, você encontrou um ponto estacionário (um pico, um vale ou um ponto de sela), onde a função não está subindo nem descendo em nenhuma direção (as derivadas parciais são todas nulas).

### A Ponte para Campos Conservativos (Mecânica e Energia)
Na física, o gradiente serve para traduzir campos de energia potencial em campos de força.
- Sempre que uma força na natureza é conservativa (como a gravidade ou a força eletrostática), ela pode ser escrita como o gradiente negativo de uma função de energia potencial $U$: $\vec{F} = -\nabla U$
- O significado físico do sinal negativo: O gradiente aponta para onde a energia potencial cresce. Como a natureza tende a buscar o estado de menor energia possível, a força real atua no sentido oposto ao gradiente ($-\nabla U$), puxando o sistema de volta para o equilíbrio.


## Divergente
$\nabla \cdot v = \frac{𝜕\upsilon_x}{𝜕x} + \frac{𝜕\upsilon_y}{𝜕y} + \frac{𝜕\upsilon_z}{𝜕z}$  $\to F(x,y,z) = P(x,y,z)\hat{x} + Q(x,y,z)\hat{y} + R(x,y,z)\hat{z}$ <br>

Exemplo: <br>

$\nabla \cdot v = x^2\hat{x} + 2y\hat{y} + 3z\hat{z}$ <br>
- Derivamos a componente $x(x^2)$ em relação a $x: \frac{𝜕}{𝜕x}(x^2) = 2x$ <br>
- Derivamos a componente $y(2y)$ em relação a $y: \frac{𝜕}{𝜕y}(2y) = 2$ <br>
- Derivamos a componente $z(3z)$ em relação a $z: \frac{𝜕}{𝜕z}3z = 3$ <br>

$\nabla \cdot v = 2x + 2 + 3 = 2x + 5$ <br>

Exemplo2: <br>

$\vec{v} = (\mathbf{x^2 y z})\hat{x} + (x y^2 z)\hat{y} + (x y z^2)\hat{z}$
- Derivamos a componente $x^2yz$ em relação a $x: \frac{𝜕}{𝜕x}(x^2yz) = 2xyz$ <br>
- Derivamos a componente $xy^2z$ em relação a $y: \frac{𝜕}{𝜕y}(xy^2z) = 2xyz$ <br>
- Derivamos a componente $xyz^2$ em relação a $z: \frac{𝜕}{𝜕z}(xyz^2) = 2xyz$ <br>

$\nabla \cdot v = 2xyz + 2xyz + 2xyz = 6xyz$ <br>

### Identificação de Fontes e Sumidouros (O Significado Geométrico)
O divergente quantifica se as setas de um campo vetorial estão "nascendo" de um ponto ou "morrendo" nele. <br>
- $\nabla \cdot \vec{v} > 0$ (Fonte): O ponto age como uma nascente. Mais linhas de campo saem do ponto do que entram (expansão líquida).
- $\nabla \cdot \vec{v} < 0$ (Sumidouro): O ponto age como um ralo. Mais linhas de campo entram no ponto do que saem (compressão líquida).
- $\nabla \cdot \vec{v} = 0$ (Campo Solenoidal): Tudo o que entra em um volume, sai do outro. Não há acúmulo nem perda local.

### A Lei de Gauss da Eletrostática
- A forma diferencial da Lei de Gauss estabelece que o divergente do campo elétrico ($\vec{E}$) em um ponto é diretamente proporcional à densidade de carga livre ($\rho$) naquele exato local: $\nabla \cdot \vec{E} = \frac{\rho}{\varepsilon_0}$
- Por que isso é poderoso? Significa que se você souber o formato espacial do campo elétrico em uma região, basta calcular o divergente dele para descobrir exatamente onde estão as cargas elétricas (as fontes do campo) que o geraram, sem precisar somar cargas uma a uma.

### Dinâmica dos Fluidos e Conservação de Massa (Mecânica)
Na mecânica dos fluidos e sistemas dinâmicos, se $\vec{v}$ representa a velocidade de um escoamento de ar ou água e $\rho$ a massa específica:
- O divergente do fluxo de massa ($\nabla \cdot (\rho \vec{v})$) mede a taxa com que o fluido está se adensando ou se diluindo em uma região infinitesimal.
- Se um fluido for incompressível (como a água em escoamentos comuns), a densidade não muda e o divergente da velocidade é estritamente zero:
$\nabla \cdot \vec{v} = 0$



# Rotacional
$\nabla \times \vec{v} = \begin{vmatrix}  \hat{x} & \hat{y} & \hat{z} \\  \frac{\partial}{\partial x} & \frac{\partial}{\partial y} & \frac{\partial}{\partial z} \\  v_x & v_y & v_z  \end{vmatrix}$

$\nabla \times \vec{v} = \left( \frac{\partial v_z}{\partial y} - \frac{\partial v_y}{\partial z} \right)\hat{x} + \left( \frac{\partial v_x}{\partial z} - \frac{\partial v_z}{\partial x} \right)\hat{y} + \left( \frac{\partial v_y}{\partial x} - \frac{\partial v_x}{\partial y} \right)\hat{z}$

Exemplo: <br>
$v_a = -y\hat{x} + x\hat{y}$ <br>
$\nabla \times \vec{v} = \begin{vmatrix}  \hat{x} & \hat{y} & \hat{z} \\  \frac{\partial}{\partial x} & \frac{\partial}{\partial y} & \frac{\partial}{\partial z} \\  -y & x & 0  \end{vmatrix}$ <br>

$\nabla \times \vec{v} = \left( \frac{\partial 0}{\partial y} - \frac{\partial x}{\partial z} \right)\hat{x} - \left( \frac{\partial y}{\partial z} + \frac{\partial 0}{\partial x} \right)\hat{y} + \left( \frac{\partial x}{\partial x} + \frac{\partial y}{\partial y} \right)\hat{z}$

$\nabla \times \vec{v} = \left( 0 \right)\hat{x} - \left( 0 \right)\hat{y} + \left( 1 + 1 \right)\hat{z} = \left( 2 \right)\hat{z}$

- Intuição Física: O "Pneu" no EscoamentoImagine que você joga uma pequena roda de pás (um cata-vento microscópico) dentro de um rio com correnteza irregular:Se a roda de pás começar a girar, o campo naquele ponto tem rotacional não-nulo ($\nabla \times \vec{v} \neq 0$). Há um cisalhamento ou um redemoinho.Se a roda de pás apenas for arrastada pela correnteza em linha reta sem girar sobre o próprio eixo, o rotacional é zero ($\nabla \times \vec{v} = 0$), mesmo que a velocidade da água varie de lugar para lugar.
- Aplicação no Eletromagnetismo (Lei de Ampère-Maxwell)Na física, o rotacional é a ferramenta perfeita para descrever campos magnéticos induzidos por correntes elétricas ou variações de campos elétricos. A forma diferencial de uma das equações de Maxwell diz que: $\nabla \times \vec{B} = \mu_0 \vec{J}$
Isso significa que o campo magnético ($\vec{B}$) "cria um redemoinho" ao redor de onde passa uma densidade de corrente elétrica ($\vec{J}$).

### Dinâmica de Fluidos e Aerodinâmica (Vorticidade)
Em mecânica dos fluidos, o rotacional mede a vorticidade — ou seja, a tendência de um fluido de girar em torno de um ponto.
- Previsão de Turbulências: Engenheiros usam o rotacional para identificar onde o escoamento ao redor de asas de aviões, pás de turbinas eólicas ou cascos de navios deixa de ser laminar e forma redemoinhos (o que gera perda de energia e arrasto).
- Sistemas Meteorológicos: Na modelagem da atmosfera, o rotacional do campo de ventos ajuda a rastrear a rotação de tempestades, frentes frias e a formação de ciclones e tornados.

### Eletromagnetismo (As Equações de Maxwell)
O eletromagnetismo moderno é construído sobre o rotacional. Duas das quatro equações fundamentais de Maxwell dependem diretamente dele:
- Lei de Ampère-Maxwell ($\nabla \times \vec{B} = \mu_0 \vec{J} + \mu_0 \varepsilon_0 \frac{\partial \vec{E}}{\partial t}$): Mostra que um campo magnético forma um "redemoinho" ao redor de correntes elétricas ou de campos elétricos que mudam com o tempo. É o princípio que faz motores elétricos e transformadores funcionarem.
- Lei da Indução de Faraday ($\nabla \times \vec{E} = -\frac{\partial \vec{B}}{\partial t}$): Demonstra que um campo magnético variável no tempo cria um campo elétrico rotacional, o que induz corrente em bobinas e geradores de energia.
